## Geologic Context

## Paleomagnetic And Rock Magnetic Data

- B = 7 sites, N = 35 samples; mean direction DEC/INC = 93.5°/7.0°; K = 48.5.
- Field/statistical tests noted in compilation: Ro.
- Rock-magnetic interpretation: expand with mineral carriers/coercivity/unblocking evidence from source studies.

## Age Constraints

- Current bracket: nominal age 1425 Ma, lomagage 1400.0 Ma, himagage 1450.0 Ma.
- Proposed new bracket: nominal age 1404 Ma, lomagage 1391 Ma, himagage 1430 Ma.
- Compilation method note: previously assigned ca.1420, The only U–Pb data reported on the Mistastin suite is an unpublished 1420 Ma date mentioned by Emslie and Stirling (1993). But a conference abstract of Mohammadi et al., 2025 reports 2 U-Pb SHRIMP zircon dates yields crystallization ages of 1404±13 Ma (monzogranite) and 1418±12 Ma (syenogranite), with a metamorphic overprint at 1401±14 Ma in the latter. We update the age bracket to 1404 Ma (nominal age), 1391 Ma (lomagage), and 1430 Ma (himagage) based on the new data. But R1 assignment remains 0.
