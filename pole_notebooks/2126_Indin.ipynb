# Indin ca. 2126 Ma paleomagnetic pole

## Geologic Context

- Terrane: Laurentia-Slave.
- Compiled unit/pole label: Indin dykes.
- The Indin diabase dykes trend northwest to north-northwest across the southern Slave craton. They cut Archean basement and the ca. 2193 Ma Dogrib dykes, and are in turn cut by the ca. 1885 Ma Ghost swarm (Buchan et al., 2016).

## Paleomagnetic And Rock Magnetic Data

- B = 20 dykes; mean direction DEC/INC = 116.7°/71.7°; K = 45.8; α95 = 4.9°; pole 40.3°N, 284.7°E, A95 = 8.0°, corrected to the common reference locality 62.5°N, 114.5°W.
- Field/statistical tests noted in compilation: C+.
- The compiled entry is the dyke-level mean of Liu et al. (2024, their Table 3), which combines the 18 dykes of Buchan et al. (2016) with 9 new sites in the central and eastern Slave craton and consolidates dykes sampled at more than one site so that oversampled dykes are not counted more than once. It comprises 4 dykes with a northwest-and-up remanence and 16 with a southeast-and-down remanence.
- Liu et al. (2024) prefer the southeast-and-down subset of 16 dykes (39.9°N, 284.4°E, A95 = 9.4°) for their paleosecular-variation calculation, to keep a possible age difference between the polarity groups from inflating the VGP scatter. The compiled entry retains both polarities. The two means are indistinguishable, lying 0.5° apart, so the position of the pole does not turn on the choice, and the case for restricting the compiled mean to one polarity does not hold up on the published record:
    - The precisely dated dyke carries an upward remanence. Buchan et al. (2016) date site I05 at 2126 ± 3 Ma on three baddeleyite fractions, and 2126 ± 2 Ma combined with a collinear fraction from site I00 on the same dyke, and state that “the dated site (I05) has an upward Indin remanence direction.” I05 is part of NWU2 in Liu et al.’s Table 3, so a southeast-and-down-only mean would exclude the polarity group that the tabulated age dates.
    - The suggestion that the swarm spans as much as 20 Myr rests on a ca. 2108 Ma U-Pb baddeleyite age reported only in abstracts (Davis and Bleeker, 2007; Bleeker et al., 2008a), from another northwest-trending dyke. This compilation does not use that date, so it is not a basis for partitioning the pole either.
- Both polarities are retained in the compiled mean, so the positive reversal test applies to the tabulated pole. Buchan et al. (2016) obtained a positive reversal test on 12 normal and 6 reversed dykes (p = 0.93); Liu et al. (2024) report a class B positive reversal test on the combined 28 sites, the 2.3° angular difference between polarities being less than the 8.3° critical angle.
- A baked contact profile across an Indin dyke cutting Archean metavolcanic rocks (Schwarz et al., 1985) shows the dyke-related overprint unblocking temperature decreasing systematically with distance from the contact, as expected of a thermal remanence acquired on cooling — a positive baked contact test that the characteristic remanence is primary.
- Rock magnetism: unblocking temperatures are sharply defined at 560 ± 10 °C, consistent with magnetite of minor titanium content as the dominant remanence carrier (Liu et al., 2024).

## Age Constraints

- Current compiled age: nominal age 2126.0 Ma, lomagage 2124 Ma, himagage 2128 Ma, from the 2126 ± 2 Ma combined upper intercept of Buchan et al. (2016).
- Compilation method note: 2126±3 or 2126±2, U-Pb baddeleyite upper intercept ages from Buchan et al., 2016; although they cite abstracts by Bleeker et al., 2008 to have a much younger date of 2108 Ma, that date is not published and we do not consider it here.
- Liu et al. (2024) add no new geochronology, so the age constraint is unchanged from Buchan et al. (2016).
