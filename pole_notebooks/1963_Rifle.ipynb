# Rifle ca. 1963 Ma paleomagnetic pole

## Geologic Context

- Terrane: Laurentia-Slave.
- Compiled unit/pole label: Rifle (Western River) Formation.

## Paleomagnetic And Rock Magnetic Data

- B = 22 sites, N = 65 samples; mean direction DEC/INC = 86.0°/25.0°; K = 12.0.
- Field/statistical tests noted in compilation: Ro,C*+ (baked contact test with intruding Mara River Sheet at 1870 Ma).
- Rock-magnetic interpretation: expand with mineral carriers/coercivity/unblocking evidence from source studies.

## Age Constraints

- Current bracket: nominal age 1963 Ma, lomagage 1957.0 Ma, himagage 1969.0 Ma.
- Compilation method note: The Rifle Formation is dated by zircons from interbedded volcanic ash layers to be 1963 ± 6 Ma (concordia upper intercept age) by ID-TIMS method by Bowring and Grotzinger, 1992 (https://doi.org/10.2475%2Fajs.292.1.1). In the original stratigraphy chronology study, the ash bed #2 was collected between the Rifle formation and the Beechey formation near Peacock Hills area. 

## References

- Paleomagnetic source: Evans,M.E., Hoye,G.S. (1981). Paleomagnetic results from the Lower Proterozoic rocks of Great Slave Lake and Bathurst Inlet areas, Northwest Territories. Geol.Surv.Canada Paper 81-10, 191-202.
- Geochronology / age-constraint note (compilation): Bowring, S. A., & Grotzinger, J. P. (1992). Implications of new chronostratigraphy for tectonic evolution of wopmay orogen, northwest canadian shield. American Journal of Science, 292(1), 1–20. https://doi.org/10.2475/ajs.292.1.1
