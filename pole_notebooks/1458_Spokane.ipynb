# Spokane pole ~1458 Ma Elston et al. (2002)

## Geologic Context

The Mesoproterozoic Belt Supergroup, located in Montana and Idaho, US, and Alberta and British Columbia, Canada, spans an enormous extent covering > 200,000 km<sup>2</sup> (Lonn et al. 2020).  It is a predominantly sedimentary succession over > 15 km thick in places, which has been subdivided into four groups. The Spokane Formation (and the correlated Grinnell Formation) are part of the Ravalli Group, which is one of the middle groups (Lonn et al. 2020).  The Belt Supergroup was an early place of paleomagnetic analyses in the 1950s and 1960s, but the majority of subsequent paleomagnetic analyses were the product of years of work by Don Elston of the United States Geologic Survey and colleagues, which were published in Elston et al. (2002).  While Elston et al. (2002) note that carbonates and green-grey-black siliclastics were analyzed, the "weak and incoherent magnetizations" led the study to focus on red beds spanning the entirety of the stratigraphic succession.

In Glacier National Park, the Grinnell Formation is a red mud-cracked mudstone and quartz-rich sandstone interpreted as playa lake and episodic sheetflood deposits (Kuhn 1987); it has been divided into five informal members. To the west and south of the park, the same unit of rocks is often termed the Spokane Formation (e.g. Harrison 1992, c.f. Winston, 2016), which is formally defined in the southeastern Helena Embayment sub-basin of the Belt Supergroup (Winston and Link, 1993). This correlation has been made due to lithologic similarities (redbeds) and depositional interpretations of subaerial exposure (Winston and Link, 1993).  Formally, the Grinnell Formation has been shown to grade westward into the Revett Formation. The Revett Formation has more quartz-rich sandstone and outcrops in western Montana through Couer d'Alene, Idaho; it has the same depositional environment interpretations as the Grinnell and Spokane Formations (Winston, 2016). In Canada, the Grinnell Formation is correlated to the Creston Formation, which also includes the overlying Empire Formation (Lonn et al., 2020). The paleomagnetic samples primarily come from the Grinnell, Creston, and Spokane Formations in the eastern portion of the Belt Supergroup.

## Paleomagnetic and rock magnetic data

Elston et al. (2022) report 23 different sites. Six of the sites are from literature data: Vitorello and Van der Voo (1977), Collinson and Runcorn (1960),  Black (1963), and Runcorn (1964). All of these sites were excluded from further calculations. The latter three studies were excluded because of study age and lab protocols, which used NRM or (in one case) blanket low-field AF demagnetization. Data from Vitorello and Van der Voo (1977) were reported as a locality mean and pass a fold test; however, these data were excluded as overprinted due to their steep inclinations (discussed further below).

Elston et al. (2002) used distinct treatments depending on the measurement lab. The USGS subjected pilot samples to stepwise demagnetization in 8 steps, but then demagnetized the remaining samples in blanket demagnetization steps--the majority of accepted sites were treated this way, with steps of 550 ºC (8 sites), 590 ºC (3 sites), and 630 ºC (1 site) used.  The Geologic Survey of Canada subjected samples to stepwise demagnetization in 8-15 steps, and directions were calculated using PCA--four sites were calculated this way. Due to the fact that these are redbeds, it appears that thermal demagnetization was typically used; however, one USGS site notes blanket demagnetization of 10 mT, and the text notes other sites that "respond to alternating field demagnetization". It is noteworthy that, due to these cleaning temperature steps, all paleomagnetic fits included both fine-grained pigmentary and coarse-grained detrital hematite (and potentially magnetite).

Of the 17 sites measured by Elston and colleagues, eight were included in calculating a formation pole.  Three sites were excluded due to scattered directions or rotation, as determined by paleomagnetism.  Elston et al. (2002) suggested that the discordant declinations in Stryker (not excluded), Avalanche Creek, and Little Prickly Pear were due to vertical-axis rotations.

The other 6 sites were excluded due to overprinting. This determination was qualitative and discussed in detail in the paper, beginning with the observation that the site mean directions had a bimodal clustering; these also did not pass a fold test, giving an impossible degree of unfolding 117% $\pm$ 7%, using Watson & Enkin (1993). Therefore, directions were separated into steep and shallow inclination groups; the former was interpreted to barely pass a fold test with 117 $\pm$ 16% untilting, and the latter passed with 105 $\pm$ 7% untilting.   Since at least two of the steeper directions were from sites along the Rocky Mountain Trench (Stryker, Le Beau Creek), which experienced a total burial of > 10 km, it was suggested that burial of 5 km during late-stage Belt Basin sedimentation led to magnetite remagnetization; thus, the steep directions were pre-tilting but still interpreted as secondary.

Elston et al. (2002) Table 1 reports 6 sites with this overprint direction that are not also rotated, whereas Table 3 reports 5. It is unclear if an additional site was excluded or if there is a typo.  Below, we have recalculated the pole based on the notes from Table 1, which includes 14 sites.

No formal rock magnetic analyses were performed; stepwise thermal demagnetization indicated that most of the remanence was lost between 500 and 675 ºC.  Post-tilting overprint directions held by magnetite were noted in a few sites.

All of the paleomagnetic sites had a normal polarity.

## Geochronology Constraints

A minimum age constraint for these units is provided by two TIMS U-Pb zircon ages from sills that intrude the Prichard and correlated Aldridge Formations: the Plains sill of 1468.8 ± 2.5 Ma (Sears et al. 1998) interpreted to intrude unconsolidated, wet sediments, and the 1468 ± 2 Ma age for a Moyie sill (which is also part of a group noted to have intruded unconsolidated sediment; Anderson and Davis 1995). The Alrdige and Prichard Formation are part of the lower Belt Group, which is the group below the Ravalli Group. 

A maximum age constraint of 1454 ± 9 Ma from SHRIMP U-Pb zircon dating is given by the overlying metabentonite from the upper Helena Formation (Evans et al. 2000). A recent LA-ICP-MS U-Pb zircon date provides a quite similar age of 1450.4 ± 7.8 Ma (Constenius et al. accepted; this is a revision of a prior age published in Hirtz et al. 2024); these overlap, and we use the Evans et al. (2000) age.  The Helena Formation is part of the Middle Belt Carbonate Group, which overlies the Ravalli Group.

The Spokane Formation falls approximately halfway between these two age constraints (1471 and 1445 Ma) so was given an age of 1458 Ma.

## Get bedding from tilt correction to do fold test

Note that data from the 1960s literature is not included here since no geographic coordinates are provided, and these sites are excluded from our preferred pole.

```python
sites_geolist = []
for site in sites_geo["site"]:
    if sites_subset2["site"].str.contains(site, regex=False).any():
        sites_geolist.append(site)

decs_fold = []
incs_fold = []
dip_dirs = []
dips = []

for site in sites_geolist:
    dec_geo = float(sites_geo.loc[sites_geo["site"] == site, "dir_dec"].iloc[0])
    inc_geo = float(sites_geo.loc[sites_geo["site"] == site, "dir_inc"].iloc[0])
    dec_tilt = float(sites_tilt.loc[sites_tilt["site"] == site, "dir_dec"].iloc[0])
    inc_tilt = float(sites_tilt.loc[sites_tilt["site"] == site, "dir_inc"].iloc[0])

    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)

    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)

diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)
```

```python
ipmag.bootstrap_fold_test(diddd)
```

The data pass a fold test. 

Bedding dips range from 9° to 56° (and one overturned sample) with the samples scattered across multiple separate fault panels/blocks.  Regional folding and faulting definitely occured in the Cretaceous but it is unclear if earlier events occurred.  Many workers suggest synsedimentary normal faulting in the lower Belt Supergroup units. There is also a proposed Kootenay orogeny (1370–1300 Ma) and widespread metamorphism documented in the Belt Supergroup, with ages around 1100 Ma (Lonn et al. 2020).

## Inclination-shallowing-corrected pole

The pole above is reported as measured. Detrital remanence in sedimentary rocks
is shallowed during compaction, so its inclination — and the paleolatitude that
inclination implies — is a minimum. Only site-level mean directions are available for this unit (15 sites), far too few for an elongation/inclination (E/I) or SVEI determination of the flattening factor.

Following Pierce et al. (2022), the flattening factor is instead resampled from
their compilation of measured *f* values (`ipmag.find_compilation_kent`). Each
resample unsquishes the mean direction implied by the pole at the sampling
locality, and the resulting distribution of mean poles is summarized as a Kent
(1982) distribution whose 95% confidence ellipse carries the flattening
uncertainty along with the uncertainty in the pole itself. This corrected pole
is exported to `data/nordic_summaries/kent/` and is the position plotted for
this unit in the compilation figures; the main compilation table keeps the
uncorrected pole.

```python
# Inclination-shallowing-corrected Kent mean pole (Pierce et al., 2022). The
# random seed is fixed so the exported summary is reproducible.
kent = pt.compilation_kent_pole(pole_subset2, study_lon, study_lat, random_seed=1)
plt.show()

Spokane_kent_summary = pt.make_kent_summary(
    Spokane_summary, kent, 'compilation', 'this study',
    comment='15 site-level mean directions: too few for an E/I or SVEI determination of f')
pt.save_kent_summary(Spokane_kent_summary, '1458_Spokane')
Spokane_kent_summary
```
