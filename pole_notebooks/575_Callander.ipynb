# Callander Complex ca. 575 Ma paleomagnetic pole



## Age constraints

- 575 ± 5 Ma U-Pb zircon date on a nepheline syenite

- Shafiqullah obtained two identical K-Ar biotite ages of 564 $\pm$ 14 Ma for the lamprophyre dikes (in Ferguson and Currie 1972). Currie (1976) gave K-Ar biotite ages of 568 Ma for a lamprophyre dike and 575 Ma for the nepheline syenite. Kamo et al. (1989b) reported a Pb-Pb isochron age using sphene, brookite, anatase, apatite, and rutile separates of 550 $\pm$ 4 Ma for a lamprophyre dike in an abstract, but they increased it to 576 $\pm$ 3 Ma on presentation. Agreement between the K-Ar and Pb-Pb ages implies cooling of the pluton within a very few million years with no later metamorphism. 

Taken together, Symons and Chasson came up with an age of 575 $\pm$ 5 Ma that appears reasonable for the Callander Complex.

- Kamo et al., 1995 report a U-Pb unpublished zircon age of 577 ± 1 Ma for the syenite but there is no published grain information for this so we stick to the 575 Ma age. 



- Shafiqullah, M., Tupper, W. M., & Cole, T. J. S. (1968). K-Ar ages on rocks from the crater at Brent, Ontario. Earth and Planetary Science Letters, 5, 148–152. https://doi.org/10.1016/S0012-821X(68)80031-7

## References

- Paleomagnetic source: Symons,D.T.A., Chiasson,A.D. (1991). Paleomagnetism of the Callander Complex and the Cambrian apparent polar wander path for North America. Canad.J.Earth Sci., 28, 355-363.
- Geochronology / age-constraint note (compilation): Age of complex is not well-constrained, K-Ar biotite ages are 564±14 and 568 Ma from lamprophyre dikes, 575 Ma for nepheline syenite, unpublished Pb-Pb age from sphene, brookite, anatase, apatite, rutile ages of 550±4/576±3 with conflicting ages between published abstract and presentation according to Symons and and Chiasson (1991). The authors assumed the age of the complex to be 575±5 Ma, but this interpretation is rather qualitative.
