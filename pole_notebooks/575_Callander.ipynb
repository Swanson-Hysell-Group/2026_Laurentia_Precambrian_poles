# Callander Complex ca. 575 Ma paleomagnetic pole

## Geologic context

The Callander alkaline (carbonatite) complex is a circular pluton ~3 km in diameter at the east end of Lake Nipissing, Ontario (Symons & Chiasson, 1991). It intrudes anorthositic and granitic gneisses of the upper Middle Proterozoic Grenville Province, and was emplaced — together with four similar alkaline complexes — along the Nipissing branch of the Ottawa–Bonnechere (St. Lawrence) graben system during Neoproterozoic extension. The complex is interpreted not to have been tilted such that the paleomagnetic data are kept in geographic (in-situ) coordinates.

## Paleomagnetic and rock magnetic data

Symons & Chiasson (1991) step-demagnetized (alternating-field and thermal) 252 specimens from 29 sites and isolated a characteristic *A magnetization component*. Isothermal-remanent-magnetization tests are interpreted to show that the A component is carried by both magnetite and hematite. A positive baked-contact test provides support for the A component being a primary ca. 575 Ma remanence. There are dual polarity directions although they do not pass a reversal test with only 3 directions in one of the modes.

The pole uses the 26 sites accepted by Symons & Chiasson (1991). Site 3 (a carbonatite dike, α95 = 25°) and site 22 (a fenite site with fewer than four reliable specimen directions) were not accepted by the authors. Site 29 is the Grenville granite gneiss of the contact test: block 29A, 0.1 m from the site-19 lamprophyre dike, is remagnetized with the dike direction, whereas blocks 29B–D, 0.6–4.1 m away, retain an older hematite-borne B component of the gneiss. The 26 accepted sites comprise 16 lamprophyre dikes, 5 syenitic-core sites (mesocratic/leucocratic nepheline syenite and potassium trachyte), and 5 fenite-aureole sites. Bringing the site VGPs to a common polarity and computing the Fisher mean reproduces the published mean direction (D = 82.2°, I = 82.7°; k = 83, α95 = 3.1°, N = 26). The Fisher mean of the site VGPs, 46.5°N, 300.8°E (A95 = 5.8°), is reported in the northern hemisphere; it is within 0.5° of the antipode (46.3°N, 301.4°E) of the published pole, 46.3°S, 121.4°E (dp = 5.9°, dm = 6.1°), which was calculated from the mean direction.

## MagIC contribution

The site data and pole are in the MagIC contribution of Symons & Chiasson (1991) prepared with this compilation (`data/575_Callander/contribution_build/build_callander.py`). It holds all site means of Table 1: the 26 accepted sites, sites 3 and 22 (with `result_quality` b), the four gneiss blocks of the contact test (not included in the pole), and the A-component pole. The B component of the gneiss is given at the site level only; the authors' B pole from the three unbaked blocks (22°S, 134°E), which they correlate with the ca. 980 Ma Grenville path, is a single-outcrop result. All sites carry the authors' locality (46.2°N, 79.4°W). Table 1 gives specimen counts only (measured, end points and remagnetization circles); these are in the `dir_n_specimens` columns, and the N of the pole is a specimen count.

## Age constraints

The age of the complex is not well-constrained in the published literature. K-Ar biotite ages are 564±14 and 568 Ma from lamprophyre dikes, 575 Ma for nepheline syenite, unpublished Pb-Pb age from sphene, brookite, anatase, apatite, rutile ages of 550±4/576±3 with conflicting ages between published abstract and presentation according to Symons and Chiasson (1991). The authors interpret the age of the complex to be 575±5 Ma, but this interpretation is rather qualitative.  Shafiqullah obtained two identical K-Ar biotite ages of 564 ± 14 Ma for the lamprophyre dikes (in Ferguson and Currie 1972). Currie (1976) reported K-Ar biotite ages of 568 Ma for a lamprophyre dike and 575 Ma for the nepheline syenite. Kamo et al. (1989b) reported a Pb-Pb isochron age using sphene, brookite, anatase, apatite, and rutile separates of 550 ± 4 Ma for a lamprophyre dike in an abstract, but they increased it to 576 ± 3 Ma on presentation. Overall, agreement between the K-Ar and Pb-Pb ages implies cooling of the pluton within a very few million years with no later metamorphism. Kamo et al., 1995 report a U-Pb unpublished zircon age of 577 ± 1 Ma for the syenite which would improve the constraints if used.

## Site-level data

The site means are loaded from this pole's MagIC `sites.txt`, built from Table 1 of Symons & Chiasson (1991) by `data/575_Callander/contribution_build/build_callander.py`. All sites are in geographic coordinates (`dir_tilt_correction = 0`; the complex is untilted). Rows flagged `result_quality` b (sites 3 and 22, not accepted by the authors) are dropped by `pt.load_magic_sites`, the A component is selected (`dir_comp_name` = characteristic), and the gneiss blocks 29A–D of the contact test are excluded, leaving the **26** accepted site means used in the pole. Table 1 gives specimen counts, so `dir_n_samples` is set from `dir_n_specimens` (205 specimens).

```python
sites_geo, sites_tc = pt.load_magic_sites('../data/575_Callander/sites.txt')
sites_geo = sites_geo[(sites_geo['dir_comp_name'] == 'characteristic')
                      & ~sites_geo['site'].astype(str).str.startswith('29')].reset_index(drop=True)
# Table 1 gives specimen counts only; N for this pole is the number of specimens
sites_geo['dir_n_samples'] = sites_geo['dir_n_specimens']
print(f'{len(sites_geo)} sites in pole')
sites_geo[['site', 'lat', 'lon', 'lithologies',
            'dir_dec', 'dir_inc', 'dir_k', 'dir_alpha95',
            'vgp_lat', 'vgp_lon',
            'dir_tilt_correction', 'dir_n_samples']]
```

All sites lie within the ~3 km circular complex at Callander Bay (46.2°N, 79.4°W); the markers overlap at this scale.

```python
pt.plot_site_map(sites_geo, zoom_start=11)
```

## Mean pole from VGPs

The Fisher mean of the 26 site VGPs (polarities unified — three sites are reversed) reproduces the published Callander pole, reported here in the northern hemisphere (~46.3°N, 301.4°E; the antipode of the published 46.3°S, 121.4°E).

```python
vgp_block, pole_mean = pt.compute_mean_pole(sites_geo, unify_polarity=True)
ipmag.print_pole_mean(pole_mean)
pole_plot = pt.plot_vgps_and_pole(vgp_block, pole_mean, central_longitude=301,
                                  central_latitude=46, figsize=(5, 5))
```
