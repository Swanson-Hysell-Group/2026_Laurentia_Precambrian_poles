# Callander Complex ca. 575 Ma paleomagnetic pole

## Geologic context

The Callander alkaline (carbonatite) complex is a circular pluton ~3 km in diameter at the east end of Lake Nipissing, Ontario (Symons & Chiasson, 1991). It intrudes anorthositic and granitic gneisses of the upper Middle Proterozoic Grenville Province , and was emplaced — together with four similar alkaline complexes — along the Nipissing branch of the Ottawa–Bonnechere (St. Lawrence) graben system during Neoproterozoic extension. The complex is interpreted not to have been tilted such that the paleomagnetic data are kept in geographic (in-situ) coordinates.

## Paleomagnetic and rock magnetic data

Symons & Chiasson (1991) step-demagnetized (alternating-field and thermal) 252 specimens from 29 sites and isolated a characteristic *A magnetization component*. Isothermal-remanent-magnetization tests are interpreted to show that the A component is carried by both magnetite and hematite. A positive baked-contact test provides support for the A component being a primary ca. 575 Ma remanence. There are dual polarity directions although they do not pass a reversal test with only 3 directions in one of the modes.

Of the 28 sites that yielded specimens, the pole uses the 26 with α95 ≤ 15°: site 3 (a carbonatite dike, α95 = 25°) and site 29 (the Grenville granite-gneiss contact-test host, α95 = 31°) are excluded, and site 22 (fewer than four reliable specimens) was not part of the digitized table. The 26 accepted sites comprise 16 lamprophyre dikes, 5 syenitic-core sites (mesocratic/leucocratic nepheline syenite and potassium trachyte), and 5 fenite-aureole sites. Bringing the site VGPs to a common polarity and computing the Fisher mean reproduces the published mean direction (D = 82.2°, I = 82.7°; k = 83, α95 = 3.1°, N = 26) and pole. The pole is reported here in the northern hemisphere at ~46.3°N, 301.4°E; this is the antipode of the published 46.3°S, 121.4°E (dp = 5.9°, dm = 6.1°) and is the same pole.

## Age constraints

The age of the complex is not well-constrained in the published literature. K-Ar biotite ages are 564±14 and 568 Ma from lamprophyre dikes, 575 Ma for nepheline syenite, unpublished Pb-Pb age from sphene, brookite, anatase, apatite, rutile ages of 550±4/576±3 with conflicting ages between published abstract and presentation according to Symons and and Chiasson (1991). The authors interpret the age of the complex to be 575±5 Ma, but this interpretation is rather qualitative.  Shafiqullah obtained two identical K-Ar biotite ages of 564 ± 14 Ma for the lamprophyre dikes (in Ferguson and Currie 1972). Currie (1976) reported K-Ar biotite ages of 568 Ma for a lamprophyre dike and 575 Ma for the nepheline syenite. Kamo et al. (1989b) reported a Pb-Pb isochron age using sphene, brookite, anatase, apatite, and rutile separates of 550 ± 4 Ma for a lamprophyre dike in an abstract, but they increased it to 576 ± 3 Ma on presentation. Overall, agreement between the K-Ar and Pb-Pb ages implies cooling of the pluton within a very few million years with no later metamorphism. Kamo et al., 1995 report a U-Pb unpublished zircon age of 577 ± 1 Ma for the syenite which would improve the constraints if used.

## Site-level data

The site means are loaded from this pole's MagIC-format `sites.txt`, assembled from Table 1 of Symons & Chiasson (1991) by `contribution_build/Callander_build_contribution.py` (there is no measurement-level MagIC contribution). All sites are in geographic coordinates (`dir_tilt_correction = 0`; the complex is untilted). The two sites with α95 > 15° (site 3 and the contact-test host gneiss, site 29) are flagged `result_quality='b'` and are dropped by `pt.load_magic_sites`, leaving the **26** accepted site means used in the pole. How the contribution is assembled is documented in `contribution_build/`.

```python
sites_geo, sites_tc = pt.load_magic_sites('../data/575_Callander/sites.txt')
print(f'{len(sites_geo)} sites in pole')
sites_geo[['site', 'lat', 'lon', 'lithologies',
            'dir_dec', 'dir_inc', 'dir_k', 'dir_alpha95',
            'vgp_lat', 'vgp_lon', 'vgp_dp', 'vgp_dm',
            'dir_tilt_correction', 'dir_n_samples']]
```

## Mean pole from VGPs

The Fisher mean of the 26 site VGPs (polarities unified — three sites are reversed) reproduces the published Callander pole, reported here in the northern hemisphere (~46.3°N, 301.4°E; the antipode of the published 46.3°S, 121.4°E).

```python
vgp_block, pole_mean = pt.compute_mean_pole(sites_geo, unify_polarity=True)
ipmag.print_pole_mean(pole_mean)
pole_plot = pt.plot_vgps_and_pole(vgp_block, pole_mean, central_longitude=301,
                                  central_latitude=46, figsize=(5, 5))
```

```python
r2 = pt.assess_R2(sites_geo, pole_mean)
print()
pt.Deenen_test(pole_mean['n'], pole_mean['alpha95'])
pt.plot_Deenen_test(pole_mean)
plt.show()
```
