# Stillwater ca. 2705 Ma paleomagnetic pole

## Geologic Context

- Terrane: Laurentia-Wyoming.

## Paleomagnetic And Rock Magnetic Data

- mean direction DEC/INC = 145.8°/59.6°; K = 31.5.
- Rock-magnetic interpretation: K-T curves, hysteresis loops, isothermal remanent magnetization (IRM) and back- eld demagnetization of IRM show that the main magnetic carriers are single-domain or pseudo-single domain magnetite.

## Age Constraints

- **nominal age 2480 Ma, lomagage 2476 Ma, himagage 2484 Ma** (legacy "2705±4" treatment).
- Concordia upper intercept age of 2480 ± 4 Ma from isotope dilution thermal ionization mass spectrometry (ID-TIMS) on four baddeleyite fractions of one Elbow Creek dike near the Stillwater Mine.
