# Western Channel diabase ca. 1592 Ma paleomagnetic pole

## Geologic context

The Western Channel diabase intrudes the folded Paleoproterozoic Rocks of the Cameron Bay and Echo Bay groups (Great Bear Magmatic Zone of Wopmay), and the generally flatlying Hornby Bay Group strata. The diabase intrudes as sheetlike bodies and dikes trending E and NNE. 

Irving et al. (1972) conducted paleomagnetic investigation of 26 sites with both drill (5-6 samples/site) and hand sample (3 samples/site). In computing the combined pole from the 35 dyke and sheet sites, they gave each site unit weight, treating every site as an independent reading of the field, which yields a small confidence circle. They cautioned, however, that some sites — the dykes in particular — may have been magnetized at essentially the same time and so do not each provide an independent estimate of secular variation, making that formal error too small. They therefore preferred a more conservative pole, calculated assuming that roughly half the sites are statistically redundant, which inflates the reported A95 from about 4° to the 6° they regarded as the best summary of the data. Hamilton and Buchan (2010) revisit this point, suggesting that combining the dyke and sheet sites adequately averages secular variation while noting that the dykes may record only two distinct VGPs. Our treatment instead recalculates the pole as the Fisher mean of all 35 site VGPs with unit weight per site (A95 = 4.3°); it does not carry Irving's redundancy inflation, so its confidence circle is correspondingly smaller than the reported 6°. 
    
## Paleomagnetic and rock magnetic data

Magnetizations were made using a spinner magnetometer, rock magnetism includes MDF of ARM, IRM. Site 17 was the only one that was unstable. 

All samples were subjected to stepwise AF demagnetization; select samples underwent thermal demagnetization (baked contact samples and Port Radium diabase site 144). High coercivity components interpreted as primary displayed unblocking temperatures between 500-560°C consistent with magnetite as the primary carrier for the diabases. No zijderveld diagrams were shown, or PCA mentioned, but the primary nature is supported by a positive baked-contact test. Irving et al. (1972) sampled fine-grained sediments and porphyritic lavas across a range of distances from the diabase: baked country rock within 8 m of contacts carries the diabase direction (Fig. 9a), an intermediate zone (15-60 m) is partially to completely remagnetized with more scattered directions (Fig. 9b), and rocks more than 1 km from contacts — including Cameron Bay Group porphyry — retain a distinct remanence unrelated to the diabase and show no evidence of reheating (Fig. 9c). Some of the near-contact porphyritic lavas were less well-behaved (high coercivity, less coherent directions), but the baked sediment contacts together with the distinct distal rocks constitute a positive baked-contact test, from which Irving et al. (1972) concluded the magnetization was acquired at the time of intrusion and cooling. Data was combined from 35 sites, resulting in a direction of 356°/-50° (a95=6), and pole of 9°N, 115°W (A95 = 6.0°, N = 35)

## Age constraints
-FA66-42 1590.2 ± 3.8 Ma (U–Pb ID-TIMS baddeleyite single grain weighted average Pb/Pb age; Hamilton and Buchan, 2010).    
-FA66-44 1591.9 ± 2.9 Ma (U–Pb ID-TIMS baddeleyite single grain weighted average Pb/Pb age; Hamilton and Buchan, 2010).    
-FA66045-6 1592.4 ± 2.5 Ma (U–Pb ID-TIMS baddeleyite fractions upper intercept age; Rogers et al., 2018)

The inverse-variance weighted mean of these three U–Pb baddeleyite dates is 1591.8 ± 1.7 Ma (2σ), adopted here (rounded to 1592 Ma) as the nominal pole age.

## R2: Techniques and statistical analysis (Meert et al., 2020)

The R2 assessment follows the three sub-criteria of Meert et al. (2020) (`resources/Meert2020_R_criteria.md`): **(a)** stepwise demagnetization by at least two methods; **(b)** component analysis by PCA or great circles; and **(c)** adequate paleosecular-variation (PSV) sampling. Sub-criterion (c) is evaluated below from the recreated site data with the Deenen et al. (2011) A95 envelope on the VGP distribution together with N ≥ 25 samples, 10 ≤ K ≤ 70, and B ≥ 8 sites; the traditional ≥ 3-samples-per-site guideline is reported but not applied as a pass/fail gate, since Sapienza et al. (2023) show that paleopole accuracy is dominated by the number of independent sites (here B = 35), with diminishing returns from additional in-site samples. Sub-criteria (a) and (b) are read from the source publication: Irving et al. (1972) isolated a stable 'cleaned' remanence by stepwise AF demagnetization with thermal checks (a is met), but the study predates principal component analysis and did not use great-circle fits (b is not met). Because component analysis (b) is not satisfied, **R2 = 0** for this pole even though the quantitative PSV-sampling sub-criterion (c) passes.

```python
r2 = pt.assess_R2(sites_geo, pole_mean)
print()
pt.Deenen_test(pole_mean['n'], pole_mean['alpha95'])
pt.plot_Deenen_test(pole_mean)
plt.show()
```

### Supplementary: VGP distribution-shape diagnostics (Fisher Q-Q and SVEI)

Beyond the formal R2(c) thresholds above, these tests examine the *shape* of the VGP distribution as a further check on PSV sampling. The Fisher Q-Q test (`ipmag.fishqq`) compares the VGP distribution against a Fisher model in its longitude (uniform) and latitude (exponential) components. The SVEI test (`pmagpy.svei`) compares the elongation (E) and elongation azimuth (V2dec) of the directional distribution at the study locality against Monte Carlo realizations of a giant-Gaussian-process field model (TK03.GAD; Tauxe & Kent, 2004) at the implied paleolatitude. These are reported as diagnostics, not as a strict reliability cutoff.

```python
fishqq_result = pt.fishqq_vgps(sites_geo, unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(sites_geo, study_lon, study_lat, model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}\u00b0")
print(f"elongation E = {svei_result['E']:.2f} "
      f"(model 95% bounds {svei_result['Esim_min']:.2f}\u2013{svei_result['Esim_max']:.2f}; "
      f"{'consistent' if svei_result['E_result'] else 'inconsistent'})")
print(f"V2 declination = {svei_result['V2dec']:.1f}\u00b0 "
      f"(model 95% bounds {svei_result['V2sim_min']:.1f}\u2013{svei_result['V2sim_max']:.1f}; "
      f"{'consistent' if svei_result['V2_result'] else 'inconsistent'})")
```

The Deenen et al. (2011) A95 envelope is satisfied (A95 = 4.3° within [2.9°, 8.7°]), and N (130), K (32), and B (35) all pass; the Fisher Q-Q test finds the VGP distribution consistent with a Fisher distribution (the uniform-longitude Mu and exponential-latitude Me statistics both fall below their critical values). Sub-criterion (c) is therefore satisfied — the few sites with only two samples do not fail it (Sapienza et al., 2023); the pole instead scores **R2 = 0** because of the absence of PCA/great-circle component analysis (sub-criterion b), as Irving et al. (1972) predates principal component analysis. The SVEI test, at the implied paleolatitude (~31°), returns an elongation magnitude (E ≈ 5.0) well above the TK03.GAD 95% range (≈1.1–3.3): the directional distribution is more elongated than secular variation alone would produce, most likely because the shallow-inclination Western Channel **dike** sites are pooled with the steeper diabase-**sheet** sites. The elongation azimuth (V2dec) is consistent with the model.
