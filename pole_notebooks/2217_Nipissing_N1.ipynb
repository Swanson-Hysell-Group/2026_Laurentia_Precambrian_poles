# Nipissing N1 ca. 2217 Ma paleomagnetic pole

## Geologic Context

- Terrane: Laurentia-Superior(East).
- Compiled unit/pole label: NIPISSING N1.

## Paleomagnetic And Rock Magnetic Data

- mean direction DEC/INC = 7.4°/-43.9°; K = 1000.0.
- Field/statistical tests noted in compilation: C+.
- Rock-magnetic interpretation: expand with mineral carriers/coercivity/unblocking evidence from source studies.

## Age Constraints

- Current bracket: nominal age 2217 Ma, lomagage 2214 Ma, himagage 2219 Ma.
- Compilation method note: A series of U-Pb dates have been developed from the Nipissing Sills in the literature, they are compiled in Table 1 of Davey et al., 2019. There is a range of ages but they are largely overlapping. We choose to use the abraded zircon U-Pb TIMS data from CA-ID-TIMS zircon and ID-TIMS baddeleyite ages from Bleeker et al., 2015 to make a call that the age constraints on the Nippising sills are between 2214 Ma and 2219 Ma.
