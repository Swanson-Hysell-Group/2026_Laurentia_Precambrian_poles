# Mistastin ca. 1441 Ma paleomagnetic pole

## Geologic Context
The Mistastin Pluton is located in the Churchill Province of eastern Quebec about 120 km north of the Grenville Front.  It is a coarse-grained, equigranular quartz monzonite batholoith. Locally, it is highly quartzose and exhibits rapakivi texture.

The Mistastin Batholith was impacted 37.9 Ma, as recorded by the Kamestastin (Mistastin) lake impact structure (Davis et al. 2025).  Paleomagnetic analyses have been performed on this impact structure and surrounding basement rocks (Hervé et al. 2015). They noted that at this location the batholith consists of granodiorite, mangerite (hypersthene monzonite), and anorthosite.

## Paleomagnetic And Rock Magnetic Data
Fahrig and Jones collected 70 cores from 11 sites in the Mistastin Batholith.  Two specimens were cut per core, and measurements were averaged to give a sample mean.  At least two specimens per site were analyzed using progressive AF demagnetization from 5 mT to 25 mT with some strong samples being subjected to additional steps of 35 mT and 45 mT.  All additional specimens were subjected to blanket cleaning at 20 mT and 25 mT and fits were performed using vector subtraction (not PCA).  No thermal demagnetization treatment was applied.

Two sites (34 & 39) did not respond to AF demagnetization, one site did had some samples that responded and others that didn't (40).  This was interpreted to be a result of recent weathering, and these samples were excluded from mean direction/pole calculatons.  One sample and two specimens within site 38 were further discarded for the same reason (and due to retaining steep directions similar to the modern field), but the site was retained. One additional site (35) had scattered directions after blanket demagnetization and was additionally excluded, leaving 7 sites to be included in the mean pole direction.  

More recently, paleomagnetic analyses have been performed on the Mistastin Lake impact structure and surrounding basement rocks (Hervé et al. 2015).  They focused on impact melts but measured 3 sites of basement rocks and one that contains both basement and melt rocks.  However, the basement-melt site and the basement site had a similar direction to the impact direction, as well as higher coercivities, unblocking temperatures, and magnetic susceptibilities, so they were presumed by the authors to be overprinted.  Samples were treated using stepwise thermal demagnetization in 12 to 14 temperature steps up to 580ºC.  In one of the remaining two sites, half of the samples were instead treated using alternating field demagnetization up to peak fields of 90 mT. All samples were analyzed using principal component analysis.

No rock magnetism was performed on these samples in Fahrig and Jones (1976). However, Hervé et al. (2015) performed hysteresis, backfield isothermal remanent magnetization, and thermomagnetic curves on the basement rocks to highlight that the main magnetic carrier at Mistastin Lake was Ti-poor magnetite with a mixture of SD and MD grains, with more MD grains as one moves away from the center of the impact structure.  Zero field cooling experiments upon warming from 80 to 130K showed a Verwey transition in the basement rocks, which got sharper as one moved away from the center of the impact structure.  Published petrography of one basement rock also showed titanomagnetite banded with ilmenite. Anisotropy analyses using ARM were also performed.  Fahrig and Jones (1976) noted a lithologic correlation with NRM intensity based on site mean comparison: specifically that pink or grey potash feldspar-rich material was weakly magnetized whereas dark greenish brown pyroxene quartz monzonite was strongly magnetized.

Dual polarities were found with one normal site and the rest reversed (as per nomenclature of paper) in the Fahrig and Jones (1976) study.  No statistical analyses are performed (see below).  2 additional normal sites were found by Hervé et al. (2015) and are included in analyses below.

This extensive pluton was sampled across a large area to try to best capture the TRM during pluton cooling. No tilt corrections were applied by either study.

Hervé et al. (2015) showed that in some locations, basement rocks were clearly overprinted by the impact, whereas others weren't. At one site (#8), some streaking between present local field and impact direction was noted; this was interpreted as a thermal overprint from the impact instead of a blanket shock impact that would affect all samples and overprint the observed viscous directions. The impact-overprinted sites, 13 and 15, were collected right below the impact melt contact (within 1 meter) versus site 8 and 16 which were more than a few meters from the contact.  Sites 8 and 16 were treated as an inverse "impact" baking test by the study to discuss how the impact affected southern sites more.

## Age Constraints

- Current bracket: nominal age 1425 Ma, lomagage 1400.0 Ma, himagage 1450.0 Ma.
- Proposed new bracket: nominal age 1441 Ma, lomagage 1419 Ma, himagage 1463 Ma.
- Compilation method note: previously assigned ca.1420, The only U–Pb data reported on the Mistastin suite is an unpublished 1420 Ma date mentioned by Emslie and Stirling (1993). But Marion and Sylvester, 2010 reported zircon LA-ICP-MS dates from four samples with concordia ages of the mangerite is 1451±12 Ma (2σ); the granodiorite gneiss is 1440±13 Ma (2σ); the anorthosite is 1438.7±8.9 Ma (2σ); and the granodiorite is 1429±10 Ma (2σ). Taken together we have the upper age of 1463 Ma, lower age of 1419 Ma, and nominal age of 1441 Ma.

## Component directions

Below we plot the directions of the data (all in geographic coordinates).  Sites from Fahrig and Jones are in blue with excluded sites in purple.   Hervé et al. (2015) sites are in red with sites overprinted by the impact in orange.

```python
plt.figure(figsize = (4,4))
ipmag.plot_net()

for row, site in sites_h25_df.iterrows():
    if site['result_quality'] == 'g':
        col = 'r'
    elif site['result_quality'] == 'b':
        col = 'orange'
    ipmag.plot_di_mean(dec=site['dir_dec'],
                       inc=site['dir_inc'],
                       a95=site['dir_alpha95'],
                       color=col)
for row, site in sites_fj76_df.iterrows():
    if site['result_quality'] == 'g':
        col = 'b'
    elif site['result_quality'] == 'b':
        col = 'purple'
    ipmag.plot_di_mean(dec=site['dir_dec'],
                       inc=site['dir_inc'],
                       a95=site['dir_alpha95'],
                       color=col)
```

We performed a common mean test of the two studies' accepted sties to make certain that the blanket AF demagnetization analytical protocol matched that of the more recent study, which used stepwise thermal or AF demagnetization.

```python
fj76_primary = sites_fj76_df[sites_fj76_df['result_quality'] == 'g']
h25_primary = sites_h25_df[sites_h25_df['result_quality'] == 'g']

di_fj76 = ipmag.make_di_block(fj76_primary['dir_dec'].tolist(), fj76_primary['dir_inc'].tolist())
di_h25 = ipmag.make_di_block(h25_primary['dir_dec'].tolist(), h25_primary['dir_inc'].tolist())

fj76_primary_flipped = pmag.flip(di_fj76, combine=True)
h25_primary_flipped = ipmag.do_flip(di_block=di_h25)

ipmag.common_mean_bootstrap(fj76_primary_flipped, h25_primary_flipped)
plt.show()
```

These fail a common means test. We will continue using the combined pole, but interpret with caution.

## Comparison with the Evans et al. (2021) compilation pole

The Evans et al. (2021) Nordic compilation adopts the Fahrig and Jones (1976) pole (GPMDB 2271: 1.0°S, 201.5°E, A95 = 7.6°, N = 7 sites), calculated from the seven primary quartz-monzonite sites. Recomputing that pole here as the Fisher mean of the seven Fahrig and Jones (1976) site VGPs reproduces the compilation value essentially exactly (1.0°S, 201.6°E, A95 = 7.7°; the first plot above), confirming that the compilation pole is faithfully recovered from the archived site data.

The pole adopted in this study (1.4°S, 205.1°E, A95 = 7.3°, N = 9 sites; the second plot above) adds the two basement sites of Hervé et al. (2015) — sites 8 and 16 — that lie more than a few metres below the impact-melt contact and escaped thermal overprinting by the ca. 36 Ma Mistastin Lake impact, so that they retain the Mesoproterozoic magnetization of the pluton. Including these two sites shifts the pole about 3.6° eastward in longitude (201.5°E → 205.1°E) and leaves the latitude essentially unchanged (1.0°S → 1.4°S). The recalculated pole lies only ~3.6° from the compilation pole, well within their mutual confidence limits, so the two are statistically indistinguishable.

We adopt the combined pole because the Hervé et al. (2015) sites were demagnetized with modern stepwise thermal and alternating-field treatment and characteristic directions isolated by principal-component analysis, corroborating and extending the alternating-field, blanket-demagnetized result of Fahrig and Jones (1976) with two additional, independently analyzed sites. The combined nine-site data set does, however, fail both the common-mean test between the two studies and the reversal test (above), so the pole is interpreted with caution and its Nordic ranking is retained at 'B'.

```python
# adopted pole (this study: Fahrig & Jones 1976 + Hervé et al. 2015 site VGPs) vs
# the Evans et al. (2021) compilation pole (GPMDB 2271 = the Fahrig & Jones 1976 pole)
compilation_plat, compilation_plon, compilation_A95 = -1.0, 201.5, 7.6
separation = float(pmag.angle([pole_mean['dec'], pole_mean['inc']],
                              [compilation_plon, compilation_plat])[0])
print(f"this study (N={int(pole_mean['n'])}):                    "
      f"{pole_mean['inc']:.1f} N, {pole_mean['dec']:.1f} E, A95 {pole_mean['alpha95']:.1f}")
print(f"Fahrig & Jones 1976 only (N={int(old_pole_mean['n'])}):        "
      f"{old_pole_mean['inc']:.1f} N, {old_pole_mean['dec']:.1f} E, A95 {old_pole_mean['alpha95']:.1f}")
print(f"Evans et al. 2021 compilation (GPMDB 2271): "
      f"{compilation_plat:.1f} N, {compilation_plon:.1f} E, A95 {compilation_A95:.1f}")
print(f"separation (this study vs compilation): {separation:.1f} deg (within the mutual A95)")
```

## Nordic workshop ranking

Prior Nordic Workshop compilations of paleomagnetic poles with expert-assigned rankings (Evans et al., 2021) have given the pole from the Mistastin Pluton a 'B' ranking. Our assessment is that careful evaluation of the pole suggests it should remain a 'B' ranking.

### Mistastin Pluton pole

The Nordic summary reports the **recalculated combined pole** (the Fisher mean of the nine Fahrig & Jones (1976) and Hervé et al. (2015) site VGPs). Recomputing the Fahrig & Jones (1976) data alone reproduces the previous Nordic compilation pole (GPMDB 2271) essentially exactly; the difference in the combined pole comes entirely from adding the two Hervé et al. (2015) basement sites.
