# Adirondack Highlands ca. 887 Ma paleomagnetic pole

## Geologic context

The Adirondack Highlands of northern New York preserve a granulite-facies metamorphic record from the Grenvillian orogeny. Brown and McEnroe (2012) collected paleomagnetic samples from an east-west traverse across the Highlands, sampling three principal rock units: (1) microcline gneisses (GMS) of the western and far-western Highlands, (2) the metamorphosed anorthosites of the Marcy Massif and outlying bodies together with associated metamorphic rocks (the other units sampled were metamorphosed syenites, pyroxene diorites and granitic gneisses), and (3) post-metamorphic fayalite-bearing granites of the Wanakena and Au Sable Forks plutons. Protracted granulite-facies Grenvillian metamorphism (T > 650 °C) persisted at high grade to ca. 996 Ma (Zhang et al., 2026) and reset pre-existing magnetizations, with new remanence acquired during subsequent slow cooling through the Tonian.

Each of the three rock units carries a stable remanence with characteristic dual polarity. The remanence is carried by ilmeno-hematite in the microcline gneisses and by magnetite in the metamorphosed anorthosites (with subsidiary hemo-ilmenite) and in the post-metamorphic fayalite granites. Brown and McEnroe (2012) reported a pole at 18.4°S, 151.1°E for the gneisses, 25.1°S, 149.0°E for the anorthosite group (combining 11 anorthosite sites with 3 associated metamorphic rock sites for a 14-site grouping in their Table 3), and 28.4°S, 132.7°E for the fayalite granite.

Zhang et al. (2026, *Tectonics*) develop new U-Pb petrochronology and thermochronology datasets from zircon, garnet, titanite, rutile, and apatite from the Adirondack Highlands. Probabilistic thermal history inversion of these data shows that the Adirondack Highlands cooled slowly throughout the Tonian, with the magnetite-carried remanence in the Marcy Massif anorthosite acquired ca. **887 ± 23 Ma** — about 85 Myr later than the ca. 970 Ma cooling-curve estimate of Brown and McEnroe (2012). Zhang et al. additionally report a positive inverse baked-contact test in which an unmetamorphosed mafic dike (site GD1) bakes its anorthosite host (site MA11); because such dikes intrude Proterozoic but not Paleozoic rocks in the region (Fisher, 1977) and are correlated with Ediacaran dikes dated elsewhere in the Grenville Province (Kamo et al., 1995; Halls et al., 2015), the test demonstrates that the anorthosite magnetite remanence is Proterozoic in age. This revised age fills a critical gap in the Laurentia apparent polar wander path between the ca. 990 Ma Jacobsville Formation pole and the ca. 780 Ma Gunbarrel pole, with implications for the position of Rodinia in the early to mid-Tonian.

## Paleomagnetic and rock magnetic data

Brown and McEnroe (2012) characterized the magnetic mineralogy of the Adirondack rocks using SEM petrography of oxide microstructures, isothermal remanent magnetization (IRM) acquisition, hysteresis loops, low-temperature susceptibility, and thermomagnetic experiments. The microcline gneisses are dominated by ilmeno-hematite with multiple generations of fine ilmenite exsolution lamellae, which give rise to the distinctive lamellar magnetism of these rocks. The metamorphosed anorthosites and the fayalite granites are dominated by magnetite (with subsidiary hemo-ilmenite in some anorthosites). The ilmeno-hematite remanence in the microcline gneisses is interpreted by Brown and McEnroe (2012) as a chemical remanent magnetization (CRM) acquired during sub-solvus exsolution of ilmenite lamellae from a titanohematite host, whereas the magnetite-carried remanence in the metamorphosed anorthosites and fayalite granites is a thermoremanent magnetization (TRM) acquired on cooling through the magnetite Curie temperature. Given current knowledge, it is more straight-forward to time calibrate the magnetite-held TRM.

Site mean directions were determined after progressive alternating-field (AF) and thermal demagnetization, with characteristic directions isolated by principal component analysis (Kirschvink, 1980). Sites with normal polarity (steep-down, SE declination) and reversed polarity (steep-up, NW declination) occur in all three rock units. Brown and McEnroe (2012) report that the normal and reversed sites of the combined anorthosites and associated metamorphic rocks do not pass the McFadden and McElhinny (1990) reversal test at the 95% level, that the microcline gneiss sites (10 reversed, 4 normal) fail it narrowly (observed angle 4.88° vs. critical angle 4.68°), and that the fayalite granite passes the test at class C (limited by only one normal-polarity site).

## Age constraints

- Current bracket: nominal age 887 Ma, lomagage 864 Ma, himagage 910 Ma.
- Zhang et al. (2026) obtain CA-ID-TIMS dates on metamorphic zircon from the Marcy Massif as young as 995.81 ± 0.74 Ma (2σ), showing that high-grade metamorphism persisted to ca. 996 Ma, and ID-TIMS rutile (890.08–850.11 Ma) and apatite (881.78–851.45 Ma) cooling dates that correlate with grain size. Their transdimensional thermal history inversion has the Marcy Massif cooling to ~570 °C at 896 +14/−8 Ma, to 550 °C at 885 +9/−7 Ma and to 525 °C at 870 +8/−6 Ma (95% CI). Taking the magnetite remanence to have been acquired between ~570 °C and ~525 °C, they give a conservative 910–864 Ma age for the anorthosite pole, presented as 887 ± 23 Ma. This bracket is a range, not a Gaussian uncertainty.
- Brown and McEnroe (2012) had estimated ca. 970 Ma by projecting a 570 °C blocking temperature onto the cooling curve of Mezger et al. (1991) for the central Highlands. Their cooling-curve ages for the other two units are ca. 960 Ma (940–960 Ma) for the microcline gneiss and ca. 990 Ma for the fayalite granite, whose emplacement is dated at 1047 ± 10 Ma (Wanakena) and 1047.3 ± 2.2 Ma (Au Sable Forks) by U-Pb zircon (McLelland et al., 2001).
- These cooling-curve estimates rest on the same kind of reasoning that put the anorthosite remanence at ca. 970 Ma, which the thermal history of Zhang et al. (2026) moves ~85 Myr younger. In the MagIC contribution the microcline gneiss is given a broad 850–960 Ma bracket, from the authors' maximum acquisition age to the youngest rutile and apatite U-Pb cooling dates of Zhang et al. (2026); the timing of ilmenite exsolution in the ilmeno-hematite (below ~520 °C, continuing to ~420 °C) is less certain than magnetite blocking. The fayalite granites are given 888–1047 Ma, from emplacement to the young limit of the time the Marcy Massif cooled through 570 °C (Zhang et al., 2026), since Brown and McEnroe (2012) argue that the granites reached the magnetite blocking temperature before the hotter metamorphic rocks of the central Highlands.

## MagIC contribution

The site data and poles are in the MagIC contribution of Brown and McEnroe (2012) prepared with this compilation (https://doi.org/10.7288/V4/MAGIC/20954; built by `data/887_Adirondack_anorthosite/contribution_build/build_brown2012_adirondack.py`). It holds the 36 site means of Table 1 (23 sites of Brown and McEnroe and 13 sites of R. Hargraves, per. comm. 2002) with the tabulated site VGPs, in one location per unit:

1. **Adirondack metamorphic anorthosite** (11 metamorphosed anorthosite and 3 associated metamorphic rock sites, the 'combined metamorphic rocks' of Brown and McEnroe; magnetite; 887 Ma, 864–910 Ma, Zhang et al., 2026), with two poles: the Fisher mean of the site VGPs, which is the pole used in this compilation, and the authors' pole calculated from the mean direction (GPMDB 9394)
2. **Adirondack microcline gneiss** (14 sites; ilmeno-hematite; 850–960 Ma), with the authors' pole (GPMDB 9393)
3. **Adirondack fayalite granite** (8 sites; magnetite; 888–1047 Ma), with the authors' pole (GPMDB 9395)

Because the remanence of these rocks was acquired on cooling after granulite-facies metamorphism, the site and pole ages in the contribution are those of cooling through the remanence acquisition temperatures (see Age constraints) rather than protolith or emplacement ages; the ca. 1159 Ma igneous zircon dates of the Marcy Massif (Zhang et al., 2026) are in its ages table.

The paper gives no site coordinates. The 11 anorthosite sites carry the positions provided by L. Brown (personal communication) and plotted in Zhang et al. (2026, Fig. 2b); the other sites carry the position of their sampling area on Fig. 1 of Brown and McEnroe (2012). The pole is calculated from the tabulated VGPs, so it does not depend on these positions. Inconsistencies in Table 1 are described under Site-level data below. The baked-contact test data of Zhang et al. (2026) are in their own MagIC contribution (https://doi.org/10.7288/V4/MAGIC/20549).

The R-criteria assessment below focuses on the 887 Ma metamorphic anorthosite pole.

## Site-level data

The 36 sites are stored once each, at `dir_tilt_correction = 0` (geographic coordinates): these basement metamorphic and igneous rocks have no paleohorizontal control and no tilt correction is applied. The MagIC `location` field assigns each site to one of the three pole groupings. The VGPs are those tabulated by Brown and McEnroe (2012); those of their AD sites follow from the site directions at a common point near 44.0°N, 76.0°W, those of the ADH sites from the directions at the estimated site positions.

Two sites of Table 1 do not follow this pattern. Table 1 prints the same direction for ADH 12 and ADH 34 (D = 240.7°, I = −73.3°); ADH 34's VGP follows from it, but ADH 12's VGP lies 17° away. The VGP of ADH 10 lies 33° from the pole calculated from its tabulated direction. In addition, the Table 1 directions do not reproduce the Table 2 means of the 11 metamorphosed anorthosites (D = 282.9°, I = −65.4°; the Table 1 directions give 278.6°, −65.9°) or of the 14 combined metamorphic rocks (283.9°, −67.3°; Table 1 gives 280.3°, −67.8°), whereas the microcline gneiss, other metamorphic rock and fayalite granite means are reproduced. Replacing the directions of ADH 10, ADH 12 or both with those implied by their tabulated VGPs does not reproduce the Table 2 means either, so the source of the discrepancy cannot be identified, and the sites are used as tabulated. The pole used here is the mean of the tabulated VGPs (−27.5°, 143.5°E, A95 = 12.5°); with VGPs calculated from the tabulated directions of ADH 10 and ADH 12 it would be −28.9°, 147.4°E (A95 = 11.5°), and with both sites excluded −26.3°, 147.8°E (A95 = 13.0°).

## Mean poles by rock unit

Each location's pole is computed as a Fisher mean of site VGPs after polarity unification (`unify_polarity=True`). All three units have a mix of normal and reversed polarity sites; the unification flips the minority polarity to the reversed, majority polarity before averaging, and the poles are reported in the southern hemisphere.

```python
for loc in ['Adirondack microcline gneiss',
            'Adirondack metamorphic anorthosite',
            'Adirondack fayalite granite']:
    sub = sites_geo[sites_geo['location'] == loc].reset_index(drop=True)
    _, pole = pt.compute_mean_pole(sub, unify_polarity=True, flip=True)
    print(f'\n=== {loc} (N={pole["n"]}) ===')
    ipmag.print_pole_mean(pole)
```

## Paleosecular variation and VGP-shape diagnostics

The Fisher-quantile (fishqq) test assesses whether the site VGPs are drawn from a
Fisher distribution, and the shape-elongation/inclination (SVEI) test compares the
VGP-scatter elongation against the TK03.GAD paleosecular-variation model expectation
at the sampling paleolatitude.

```python
fishqq_result = pt.fishqq_vgps(sites_anorth, unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(sites_anorth, study_lon, study_lat, model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}°")
print(f"elongation E = {svei_result['E']:.2f} "
      f"(model 95% bounds {svei_result['Esim_min']:.2f}–{svei_result['Esim_max']:.2f}; "
      f"{'consistent' if svei_result['E_result'] else 'inconsistent'})")
print(f"V2 declination = {svei_result['V2dec']:.1f}° "
      f"(model 95% bounds {svei_result['V2sim_min']:.1f}–{svei_result['V2sim_max']:.1f}; "
      f"{'consistent' if svei_result['V2_result'] else 'inconsistent'})")
```

The Fisher Q-Q test is consistent with a Fisher distribution of the VGPs. The SVEI elongation (E) is consistent with the TK03.GAD field model at the implied paleolatitude, while the elongation azimuth (V2dec) falls just outside the model bounds. With only N = 14 site VGPs these distribution-shape statistics are sensitive to small numbers and are reported as diagnostics, not as a reliability cutoff.
