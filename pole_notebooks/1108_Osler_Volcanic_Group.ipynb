# Osler Volcanic Group ca. 1108–1095 Ma paleomagnetic poles

## Geologic context

The Osler Volcanic Group is among the earliest volcanic successions of the North
American Midcontinent Rift, erupted during the rift's *ca.* 1110–1105 Ma "early
magmatic stage" as plateau tholeiitic basalts that spread over a broad
geographic area before the central Lake Superior rift graben fully developed
(Swanson-Hysell et al., 2014b). The group overlies the epicontinental sandstones
of the Mesoproterozoic Sibley Group; its lowest ~100 m contains rift-related
sandstones and conglomerates, above which is a thick, relatively continuous
succession of tholeiitic basalt lava flows.

The stratigraphically-resolved record was sampled along the east shore of
Simpson Island in the Lake Superior Archipelago, where slightly over
3000 m of Osler stratigraphy is exposed in well-preserved flows. The flows
have a median thickness of ~5 m and are dominantly subaerial pahoehoe basalts,
showing the characteristic up-flow transition from basal pipe vesicles, through
massive basalt, to amygdaloidal and flow-banded tops. Thin interflow red-brown
siltstone and basaltic-pebble conglomerate beds preserve paleohorizontal and were
used for the structural (tilt) corrections of the paleomagnetic data. The sampled
section lies more than 10 km from the felsic St. Ignace Island Complex and is
interpreted to be unaffected by its late-rift emplacement; the flows are tilted
toward the rift so that stratigraphically higher flows crop out farther south.

The Osler Group spans a major part of the Midcontinent Rift's polarity-zone
sequence, but not as a single continuous reversal. The reversed flows belong
to the early Alona Bay reversed-polarity zone (the oldest rift polarity zone),
whereas the normal flows above the unconformity belong to the much younger
Portage Lake normal-polarity zone. The two intervening Flour Bay normal and
Flour Bay reversed polarity zones — recorded elsewhere in the rift, e.g. at
Mamainse Point — are *not* preserved at Osler; they fall within the time that is
missing at the angular unconformity (Swanson-Hysell et al., 2019). To the
southwest, in the Nipigon Strait region (Puff Island), that angular
unconformity, marked by a conglomerate, separates the underlying Alona Bay
reversed flows from the overlying Portage Lake normal flows (Halls, 1974); only
~110 m of normally magnetized flows are exposed above it before the sequence
passes beneath Lake Superior. The unconformity records a period of volcanic
quiescence interpreted as the "latent stage" of rift development. The abrupt
juxtaposition of two non-adjacent polarity zones across this hiatus — rather than
an in-situ reversal — reflects the substantial missing time. This juxtaposition does not constitute a
positive unconformity test: the formal unconformity test of Kirschvink (1978)
applies to multiple magnetostratigraphic polarity zones truncated beneath a
dipping unconformable surface, whereas at Osler a single reversed zone below the
unconformity is overlain by a single, non-matching normal zone above it. A
polarity change across an unconformity records the missing time but does not by
itself field-test the age of the magnetization.

Three studies contribute the site-level data analyzed here: Halls (1974)
sampled reversed and normal flows close to the unconformity in the Nipigon Strait
region; Swanson-Hysell et al. (2014b) developed the stratigraphically-resolved
reversed-polarity record through the Simpson Island section; and Swanson-Hysell
et al. (2019) added reversed flows at Agate Point and the normal flows at Puff
Island, together with a U-Pb date.

## Pole context

Four poles are computed here, presented from oldest to youngest:

1. Osler reverse lower (R1) — *ca.* 1108 Ma; lower Alona Bay reversed zone
   (Simpson Island flows below ~1041 m). GPMDB 9515.
2. Osler reverse middle — *ca.* 1107 Ma; transitional reversed flows
   (Simpson Island 1041–2082 m). GPMDB 10017.
3. Osler reverse upper (R2) — *ca.* 1105 Ma; upper Alona Bay reversed zone
   (Simpson Island flows above ~2082 m + Halls (1974) reversed + Agate Point),
   the best-determined Osler pole. GPMDB 9514.
4. Osler normal — *ca.* 1095 Ma; the overlying Portage Lake normal-polarity
   zone (two Halls (1974) normal flows + two Puff Island flows).

The three reversed-polarity poles are the lower/middle/upper-third Osler reversed
poles of the Nordic compilation and are exported to the Nordic summaries. The
Osler normal pole is computed and shown for context (and kept in the
contribution's locations table), but is not included in the Nordic compilation
because it has too few sites (only four cooling units; N = 4).

## Paleomagnetic and rock magnetic data

The characteristic remanence is carried by magnetite and isolated by stepwise
alternating-field and thermal demagnetization with principal-component analysis;
site means are reported in tilt-corrected coordinates. A positive fold test on
the upper reversed-polarity data combining Halls (1974) and Swanson-Hysell et al.
(2014b) (Tauxe & Watson, 1994; McFadden & McElhinny, 1990) establishes a primary,
syn-rift origin for the magnetization. The lower reversed-polarity flows are
additionally crosscut by younger normal-polarity (Portage Lake) dikes, indicating
they have not been remagnetized.

## Age constraints

- A rhyolite at Agate Point yields a U-Pb (<sup>206</sup>Pb/<sup>238</sup>U)
  date of 1105.15 ± 0.33 Ma (Swanson-Hysell et al., 2019), dating the upper
  reversed-polarity zone (R2).
- The lower and middle reversed-polarity zones are older than 1105.15 Ma, with
  best estimates of *ca.* 1108 Ma (R1) and *ca.* 1107 Ma (middle), consistent
  with the *ca.* 1109 ± 1 Ma Logan sills (Davis & Sutcliffe, 1985).
- The normal-polarity flows post-date the unconformity; their age is poorly known
  (*ca.* 1095 Ma, plausibly 1080–1100 Ma).

## Site-level data and the four pole groups

## Osler reverse lower (R1) — lower Alona Bay reversed-polarity zone (ca. 1108 Ma)

Simpson Island reversed flows below ~1041 m (Swanson-Hysell et al., 2014b), N = 30; the oldest Osler pole.

```python
sub = sites_tc[sites_tc['location'] == 'Osler reverse lower'].reset_index(drop=True)
study_lat, study_lon = LOCALITY['Osler reverse lower']
vgp_block, pole_mean = pt.compute_mean_pole(sub, unify_polarity=True)
ipmag.print_pole_mean(pole_mean)
_ = pt.plot_vgps_and_pole(vgp_block, pole_mean, central_longitude=pole_mean['dec'],
                          central_latitude=pole_mean['inc'], figsize=(5, 5))
```

```python
dir_block, dir_mean = pt.compute_mean_direction(sub, unify_polarity=False)
ipmag.print_direction_mean(dir_mean)
ipmag.plot_net()
ipmag.plot_di(di_block=dir_block, color='blue', marker='o')
ipmag.plot_di_mean(dir_mean['dec'], dir_mean['inc'], dir_mean['alpha95'], color='red', marker='s')
plt.title('Osler reverse lower — site mean directions (tilt-corrected)')
plt.show()
results['Osler reverse lower'] = dict(sub=sub, pole=pole_mean, dir=dir_mean,
                        study_lat=study_lat, study_lon=study_lon)
```

```python
ovg_lower = results['Osler reverse lower']
pt.Deenen_test(ovg_lower['pole']['n'], ovg_lower['pole']['alpha95'])
pt.plot_Deenen_test(ovg_lower['pole'])
```
