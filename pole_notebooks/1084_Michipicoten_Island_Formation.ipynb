# Michipicoten Island Formation ca. 1084 Ma paleomagnetic pole

## Geologic context

Michipicoten Island, in northeastern Lake Superior (Ontario), exposes some of the youngest volcanic rocks of the North American Midcontinent Rift. Following Annells (1974), the island records three magmatic intervals: (1) a basal sequence of subophitic to ophitic olivine-tholeiitic basalt flows (the Quebec Mine Member); (2) hypabyssal quartz porphyry, granophyre, and basaltic-andesite intrusions emplaced into the Quebec Mine Member, which dominate the northern half of the island; and (3) the Michipicoten Island Formation, which unconformably overlies the intrusions (clasts of the intrusive phases occur in basal agglomerate and in the overlying West Sand Bay tuff) and forms the uppermost units of the island. The flows and tuffs dip southward toward the rift axis, more steeply (\~40°) in the older northern volcanics than in the south (\~15–20°).

The Michipicoten Island Formation consists, from bottom to top, of: glassy glomerophyric andesite flows of the Cuesta Member (\~300 m); volcaniclastic sandstone overlain by an andesite flow of the Channel Lake Member (\~270 m); green lithic tuff of the West Sand Bay Member (\~15 m); thick andesite flows of the Quebec Harbour Member (\~270 m); laminated tuff and siltstone (\~30 m) overlain by a sequence of relatively thin, olivine-free tholeiitic basalt flows of the South Shore Member (\~220 m); and, as the uppermost exposed flow, the pink rhyolite of the Davieaux Island Member. A thin carbonate bed and quenched flow tops in the easternmost South Shore Member outcrops indicate that some eastern flows erupted into a lacustrine environment.

## Paleomagnetic and rock magnetic data

The pole combines new site means from the South Shore Member tholeiitic basalt flows of Fairchild et al. (2017) — the member with the most individual cooling units — with site means from the Quebec Harbour and Davieaux Island Members reported by Palmer & Davis (1987). Stepwise thermal demagnetization with principal-component analysis isolates a characteristic remanence. Magnetizations unblock both below \~580°C (consistent with a (titano)magnetite carrier) and above \~580°C (hematite); significant unblocking between 580–625°C suggests a maghemite-bearing, chemically modified thermal remanence. Within samples and sites that carry both mineralogies the remanence directions are indistinguishable (mean difference between low- and high-temperature fits \~2.9°). Site means are westerly with shallow positive inclination and are of a single (normal) polarity.

The sampled flows are stratigraphically bracketed by the two dated horizons — the West Sand Bay Member tuff below and the Davieaux Island Member rhyolite above — so the pole is directly bracketed in age by the U-Pb geochronology.

Following the careful, date-bracketed selection of Fairchild et al. (2017), the pole avoids multiple VGPs from a single cooling unit: the 21 South Shore Member basalt sites (high-temperature, tilt-corrected) are combined with two Palmer & Davis (1987) cooling-unit means — the Quebec Harbour Member (their site means 2 and 4) and the Davieaux Island Member (their site means 15 and 17). The lower Cuesta and Channel Lake Members (below the West Sand Bay tuff) and Palmer's own South Shore sites are excluded. The Fisher mean of these 23 site VGPs gives a pole at 17.0°N, 174.7°E (A95 = 4.4°, k = 47.9, N = 23), reproducing the pole of Fairchild et al. (2017).

## Age constraints

- The sampled part of the formation is bracketed by U-Pb (CA-ID-TIMS, <sup>206</sup>Pb/<sup>238</sup>U) zircon dates of 1084.35 ± 0.20 Ma (West Sand Bay Member tuff) and 1083.52 ± 0.23 Ma (Davieaux Island Member rhyolite) (Fairchild et al., 2017). These are the youngest dates yet obtained from Midcontinent Rift volcanic rocks; the Davieaux Island rhyolite may be the youngest exposed lava flow in the rift.

## Nordic workshop ranking

Prior Nordic Workshop compilations assign the Michipicoten Island Formation an 'A' ranking — its precise U-Pb calibration and high-quality directional data make it a key Keweenawan Track tie point despite an R-score of 5/7 (the failing criteria are R4, no field test constraining the magnetization age, and R6, single polarity). Recreating the pole from the site level following Fairchild et al. (2017)'s careful, date-bracketed selection (N = 23) reproduces their published pole (17.0°N, 174.7°E) and supports retaining 'A'.
