# Marathon R ca. 2104 Ma paleomagnetic pole

## Geologic Context

- Terrane: Laurentia-Superior(West).
- Compiled unit/pole label: MARATHON R.

## Paleomagnetic And Rock Magnetic Data

- mean direction DEC/INC = 134.4°/-56.4°; K = 1000.0.
- Field/statistical tests noted in compilation: C+.
- Rock-magnetic interpretation: expand with mineral carriers/coercivity/unblocking evidence from source studies.

## Age Constraints

- Proposed new bracket: nominal age 2106 Ma, lomagage 2102.0 Ma, himagage 2110.0 Ma.
- - Current bracket: nominal age 2104 Ma, lomagage 2101.0 Ma, himagage 2106.0 Ma.
- Compilation method note: Hamilton et al., 2002, Halls et al., 2005, and Halls et al., 2008 both did U-Pb baddeleyite dating on the Marathon dikes, the findings are that the R polarity dikes appeared o have younger ages than the N polarity dikes. There are a few dated R polarity dikes that show ages around 2104, 2106, 2106 Ma, whereas there is a N polarity dike that yielded an age of 2125.7 $\pm$ 1.2 Ma. All ages are single crystal baddeleyite TIMS dates.
