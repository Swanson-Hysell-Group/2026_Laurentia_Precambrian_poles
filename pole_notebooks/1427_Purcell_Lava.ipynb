# Purcell Lava ca. 1427 Ma paleomagnetic pole

## Geologic Context

The Mesoproterozoic Belt-Purcell Supergroup, located in Montana and Idaho, USA and Alberta and British Columbia, Canada spans an enormous extent covering > 200,000 km<sup>2</sup> (Lonn et al. 2020).  It is a predominantly sedimentary succession over > 15 km thick in places, which has been subdivided into four groups. The Purcell Lava intrudes between the Shepard and Snowslip Formations, which are part of the Missoula Group (Lonn et al. 2020).  The Belt-Purcell Supergroup was an early place of paleomagnetic analyses in the 1950s and 1960s, but the majority of subsequent paleomagnetic analyses were the product of years of work by Don Elston of the United States Geologic Survey and colleagues, which were published in Elston et al. (2002).  

The Purcell Lava consists predominaly of alkalic basalts and extends in patches across northern Montana from near Libby, Montana (west), where it reaches a maximum of 385 m thick, to thinner exposures in Glacier National Park (east). In Canada, it is correlated to the Nicol Creek Formation and ranges in thickness from a few tens of meters to 750 m thick, spreading across East Kooteney, British Columbia, to Prairie Bluff, Alberta (McMechan 1981, Constenius et al. accepted).  All areas of the Purcell Lava consist of basal pillow basalts (subaqueous) overlain by pahoehoe flows (interpreted subaerial). Central and western exposures have distinct porphyritic flows with large feldspar crystals.  West of the Whitefish-Galton Range, the Purcell Lava consists of two sets of flow assemblages separated by a middle section of argillite, siltite, and sandstone (McMechan 1981, Constenius et al. accepted).  It is interpreted that the eastern exposures represent the lower Purcell Lava only.

## Paleomagnetic and Rock Magnetic Data

Elston et al. (2002) reports 12 paleomagnetic sites of the Purcell Lava and Nicol Creek Formation.  Descriptions in the text suggest that Elston's sites might include more than one flow unit (samples separated by 40 m). Two of the sites noted above are from preexisting literature but reported by Elston.  Evans et al. (1975) is included as a locality mean of three sites; this study sampled a 50 m section with sites evenly spaced ~20 m apart and notes that the uppermost site was highly vesiculated, suggesting it was near a flow top.  There were no geologic indicators that the section represented more than one flow; however, we have included these sites below to probe whether the data are distinct or share a common mean, since these site means were published in the original manuscript. Another site from the literature (Black 1963) is reported but not included in the formation mean, as it used outdated lab protocols such as blanket low-field AF demagnetization.

No separate rock magnetic analyses were done; however, reddened flow tops and microscopically observed hematite pigment on the subaerial flows suggest hematization by exposure to the atmosphere during and shortly after cooling (R. A. Burwash, pers. comm. 1974, reported in Evans et al. 1975).  Stepwise thermal demagnetization experiments were interpreted to indicate the main carrier is hematite. Examination of the Zijdervelds shows continual progressive demagnetization with larger losses between 200 - 400 ºC (pyrrhotite?) and 600-675 ºC (especially 650-675 ºC), suggesting the presence of hematite, potentially including both fine-grained (pigmentary) hematite and coarser grained specularite. Magnetite and pyrrhotite-bearing remanences were also noted, including in samples bearing hematite; the displayed Zijderveld shows an example with large losses by 300 ºC and near-complete demagnetization by 600 ºC. Sampled pillow basalts were interpreted to lack hematite and retained unreliable paleomagnetic records. 

Elston et al. (2002) used distinct treatments depending on the measurement lab. The USGS subjected pilot samples to stepwise demagnetization in 8 steps, but then demagnetized the remaining samples in blanket demagnetization steps--two sites were treated this way with a single step of 550 ºC. The Geologic Survey of Canada subjected samples to stepwise demagnetization in 8-15 steps, and directions were calculated using PCA. It appears only thermal demagnetization was used, due to the known presence of hematite, but it is not clearly reported.  Evans et al. (1975) performed stepwise demagnetization with one sample per site using AF and one using thermal.  However, the analyses were all performed based on vectors to endpoints from a single step that was chosen to maximize clustering.

Of the 12 paleomagnetic sites, three sites (Blacktail Creek, Rexford, Teepee Creek) were separated into normal and reversed directions (following Elston's nomenclature) to provide distinct site means.  All of these normal directions were excluded from the formation mean due to "contamination" or "overprinting", as the directions were not antipodal, although no formal statistical reversal test was reported.  The study also noted that these directions had steeper inclinations and were thus grouped with two other sites with steep inclinations (Sage Creek and South Lost Creek).  These sites were also separated into two groups, with the group with steeper directions being excluded due to complete or partial overprint. The interpreted overprinted directions were usually in samples that had unblocking temperatures < 600 ºC, whereas the accepted samples had main remanence carried at > 650 ºC. One final site (Bull River) was excluded due to apparent rotation, based on the paleomagnetic data.

Elston et al. (2002) provided a formation mean and VGP for the interpreted primary direction.  It passed a fold test using 100 ± 4 % untilting as the optimum, following Watson and Enkin (1993).  The study also provided a formation mean and VGP for the steeper inclination directions from 5 sites, interpreted as an overprint.  This direction had optimum untilting at 82 ± 10 % using Watson and Enkin (1993), suggesting it does not pass a fold test.

## Geochronology constraints

Evans et al. (2000) dated a restricted "porphyritic rhyolite to quartz latite flow" from the lower part of the upper Purcell lava flows outcropping on Robinson Mountain near Libby, Montana, USA; using SHRIMP U–Pb zircon dating, the study got an age of 1443 ± 7 Ma.

Constenius et al. (accepted) revisited the same location and reinterpreted it as a peperite. Based on detrital zircon U-Pb dating, they suggest the prior age is not a maximum depositional age.  Using LA-ICP-MS U-Pb zircon dating on the youngest population, the study reports an age of 1419.9 ± 8.4 Ma.  They also did ID-TIMS U-Pb dating on baddeleyite from the lower Purcell and report an age of 1438.9 ± 3.2 Ma.  Dating of instrusive sills using the alternative dating techniques corroborate magmatism at these ages in the Belt-Purcell Supergroup.

Overall, due to the lack of documentation on which flow units were sampled for paleomagnetism, we choose to use these recent ages as the mimumum and maximum constraints (with error 1412-1442 Ma), as this range also incorporates the older age.  The nominal age is the middle at 1427 Ma.

# Recalculate pole based on VGPs

# Get bedding from tilt correction to do fold test

Elston et al. (2002) reported bedding directions for each site; however, we have calculated these directions based on the reported data to perform a fold test.

```python
# Find sites that have both geographic and tilt-corrected coordinates

sites_geolist = []
for site in sites_geo['site']:
    if sites_tilt['site'].str.contains(site).any():
        sites_geolist.append(site)

decs_fold = []
incs_fold = []        
dip_dirs = []
dips = []
for site in sites_geolist:
    dec_geo = float(sites_geo[sites_geo['site'] == site]['dir_dec'])
    inc_geo = float(sites_geo[sites_geo['site'] == site]['dir_inc'])
    dec_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_dec'])
    inc_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_inc'])
    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)
    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)
    
diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)

ipmag.bootstrap_fold_test(diddd)
```

These data (including sites excluded by Elston) do pass a fold test.

We have also performed a fold test with only the sites Elston et al. (2002) included in the primary VGP.

```python
# Filter for Elston's subset
sites_subset = sites_tilt[sites_tilt['result_quality'] == 'g']
sites_subset_geo = sites_geo[sites_geo['result_quality'] == 'g']

# Find sites that have both geographic and tilt-corrected coordinates
sites_geolist = []
for site in sites_subset_geo['site']:
    if sites_subset['site'].str.contains(site).any():
        sites_geolist.append(site)

decs_fold = []
incs_fold = []        
dip_dirs = []
dips = []
for site in sites_geolist:
    dec_geo = float(sites_subset_geo[sites_subset_geo['site'] == site]['dir_dec'])
    inc_geo = float(sites_subset_geo[sites_subset_geo['site'] == site]['dir_inc'])
    dec_tilt = float(sites_subset[sites_subset['site'] == site]['dir_dec'])
    inc_tilt = float(sites_subset[sites_subset['site'] == site]['dir_inc'])
    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)
    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)
    
diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)

ipmag.bootstrap_fold_test(diddd)
```

These data do pass a fold test.

## Nordic workshop ranking

Prior Nordic Workshop compilations of paleomagnetic poles with expert-assigned rankings (Evans et al., 2021) have given the pole from the Purcell Lava an 'A' ranking. 

Our assessment and community evaluation by the 2026 Nordic Workshop is that careful evaluation of the pole suggests it should be assigned a 'B' ranking, primarily due to the poorly documented mineralogical complexity which make R3 and R6 scores of 0.
