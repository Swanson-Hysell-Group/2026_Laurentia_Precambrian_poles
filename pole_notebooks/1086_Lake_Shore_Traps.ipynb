# Lake Shore Traps ca. 1085 Ma paleomagnetic pole

## Geologic context

The Lake Shore Traps are a sequence of subaerial tholeiitic basalt lava flows interbedded within the upper Copper Harbor Conglomerate along the northern Keweenaw Peninsula of Michigan. They record a late, post-Portage-Lake-Volcanics pulse of effusive magmatism in the North American Midcontinent Rift and were emplaced during the normal-polarity ("Portage Lake") interval of rift volcanism. The flows now dip toward the rift axis as part of the Keweenaw homocline, with bedding attitudes that vary between sampling localities.

Site-level paleomagnetic data come from two studies: Diehl & Haig (1994) reported flow-by-flow site means from the Middle and Outer members, and Kulakov, Smirnov & Diehl (2013) re-sampled the succession and added new sites (Schroeder Creek and additional Lake Shore localities) together with a paleointensity and rock-magnetic study. The flow LST28 was measured independently by both studies; it is counted once here so that each cooling unit contributes a single site mean.

## Paleomagnetic and rock magnetic data

Stepwise thermal and alternating-field demagnetization isolate a characteristic remanent magnetization carried by low-titanium magnetite (maximum unblocking temperatures near 580°C). Site mean directions are westerly with moderate positive (downward) inclination in tilt-corrected coordinates, consistent with the normal-polarity Portage Lake interval. Kulakov et al. (2013) document the rock magnetic behavior in detail in support of their Thellier-type paleointensity determinations.

Combining the two studies and counting the duplicated flow once yields 50 site means. Bringing the site VGPs to a common polarity and computing the Fisher mean gives a pole at 23.1°N, 185.7°E (A95 = 3.3°, N = 50), in agreement with the previously compiled Lake Shore Traps pole.

## Age constraints

- An andesitic flow within the Lake Shore Traps yields a U-Pb (206Pb/238U, CA-ID-TIMS) weighted-mean date of 1085.57 ± 0.5 Ma (Fairchild et al., 2017), which we adopt as the best age for the pole.
- Earlier work assigned the Lake Shore Traps an age of 1087.2 ± 1.6 Ma based on U-Pb dating of an interflow andesite (cited by Kulakov et al., 2013).
- The flows are stratigraphically above the bulk of the Portage Lake Volcanics, consistent with an age slightly younger than the ca. 1091 Ma upper Portage Lake sequence.

## R2: Techniques and statistical analysis (Meert et al., 2020)

The R2 assessment follows the three sub-criteria of Meert et al. (2020), as set out in `resources/Meert2020_R_criteria.md`: **(a)** stepwise demagnetization by at least two methods (AF and thermal); **(b)** component analysis by PCA (best-fit lines) or great circles; and **(c)** adequate paleosecular-variation (PSV) sampling. Sub-criteria (a) and (b) are reported here only as *advisory* indicators read from the MagIC `method_codes`, which are inconsistently applied between studies; the two-demag-method requirement in particular is not universally expected (e.g. AF is inappropriate for hematite-bearing rocks) and is confirmed against the source publication. The substantive, reproducible part of R2 is sub-criterion (c), evaluated from the recreated site data with the Deenen et al. (2011) A95 envelope on the VGP distribution together with N ≥ 25 samples, 10 ≤ K ≤ 70, B ≥ 8 sites, and ≥ 3 samples per site.

```python
r2 = pt.assess_R2(sites_tc, pole_mean)
print()
pt.Deenen_test(pole_mean['n'], pole_mean['alpha95'])
pt.plot_Deenen_test(pole_mean)
```

### Supplementary: VGP distribution-shape diagnostics (Fisher Q-Q and SVEI)

Beyond the formal R2(c) thresholds above, these tests examine the *shape* of the VGP distribution as a further check on PSV sampling. The Fisher Q-Q test (`ipmag.fishqq`) compares the VGP distribution against a Fisher model in its longitude (uniform) and latitude (exponential) components. The SVEI test (`pmagpy.svei`) compares the elongation (E) and elongation azimuth (V2dec) of the directional distribution at the study locality against Monte Carlo realizations of a giant-Gaussian-process field model (TK03.GAD; Tauxe & Kent, 2004) at the implied paleolatitude. These are reported as diagnostics, not as a strict reliability cutoff.

```python
fishqq_result = pt.fishqq_vgps(sites_tc, unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(sites_tc, study_lon, study_lat, model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}\u00b0")
print(f"elongation E = {svei_result['E']:.2f} "
      f"(model 95% bounds {svei_result['Esim_min']:.2f}\u2013{svei_result['Esim_max']:.2f}; "
      f"{'consistent' if svei_result['E_result'] else 'inconsistent'})")
print(f"V2 declination = {svei_result['V2dec']:.1f}\u00b0 "
      f"(model 95% bounds {svei_result['V2sim_min']:.1f}\u2013{svei_result['V2sim_max']:.1f}; "
      f"{'consistent' if svei_result['V2_result'] else 'inconsistent'})")
```

The SVEI test on the directions finds an elongation magnitude (E ≈ 2.7) consistent with the TK03.GAD field model at this paleolatitude, but an elongation azimuth (V2dec) that departs from the model expectation. Correspondingly, the Fisher Q-Q test on the VGPs rejects a Fisher distribution — this is driven by a non-uniform longitude ('declination') component (Mu above its critical value), while the latitude component is essentially consistent. VGPs that adequately average paleosecular variation are generally expected to be approximately Fisherian (close to circularly symmetric about the mean), so these results indicate that the VGP distribution about the mean has a somewhat unusual shape rather than ordinary symmetric PSV scatter. They are reported here as distribution-shape diagnostics.

(2013) data (50 unique cooling-unit site means) reproduces the previously compiled pole position and supports retaining the 'A' ranking. The high-precision U-Pb andesite date of Fairchild et al. (2017) provides a tighter age (1085.57 ± 0.5 Ma) than the value carried in earlier compilations.
