# Snowslip pole ~1449 Ma Elston et al. (2002)

## Geologic Context

The Mesoproterozoic Belt Supergroup, located in Montana and Idaho, USA, and Alberta and British Columbia, Canada, spans an enormous extent covering > 200,000 km<sup>2</sup> (Lonn et al. 2020).  It is a predominantly sedimentary succession over 15 km thick in places, which has been subdivided into four groups. The Snowslip Formation is in the Missoula Group, which is the uppermost group (Lonn et al. 2020).  The Belt Supergroup was an early place of paleomagnetic analyses in the 1950s and 1960s, but the majority of subsequent paleomagnetic analyses were the product of years of work by Don Elston of the United States Geologic Survey and colleagues, which were published in Elston et al. (2002).  While Elston et al. (2002) note that carbonates and green-grey-black siliclastics were analyzed, the "weak and incoherent magnetizations" led the study to focus on red beds spanning the entirety of the stratigraphic succession.

The Snowslip Formation was defined in Glacier National Park, Montana, and contains three red mudcracked mudstone and quartz-rich sandstone units, separated by two green units of interbedded siltstone and mudstone with fewer mudcracks (Winston and Link 1993).  It has been interpreted as intertidal to peritidal, with episodes of sheetflood deposits (Winston and Link 1993).  The Snowslip Formation is interpreted to extend across western Montana to Clark Fork and Bonner's Ferry, Idaho, through correlation with the lower portion of the Upper Wallace Formation, which is less mudcracked and sandstone-rich (Lonn et al. 2020).  The Upper Wallace and Snowslip Formation are correlated with the Van Creek Formation in Canada (Lonn et al. 2020), although all of Elston et al. (2002)'s samples are from the United States and were noted to be from the base of the Snowslip Formation.

## Paleomagnetic and rock magnetic data

Elston et al. (2002) report 13 sites from the Snowslip Formation, which includes two sites from the literature.  Collinson and Runcorn (1960) measured an NRM, while Black (1963) used blanket demagnetization to 16 mT respectively. While Table 1 does not note that they were excluded from the formation mean, the caption notes that all 1960s studies were excluded.  Additionally, in Table 3, the included number of sites used in the mean is 2 less than listed in Table 1 (and mean pole calculations suggest that excluding the 1960s sites is closest to the reported mean). A formation mean of 9 sites is reported and calculated using Fisher statistics.

The USGS subjected pilot samples to stepwise demagnetization in 8 steps, but then demagnetized the remaining samples in blanket steps -- in this case a 550 ºC step was used (which would include magnetite and all forms of hematite).  The single stepwise thermal demagnetization Zidjerveld shown highlighted that the vast majority of remanence is held at > 650 ºC, suggesting the presence of detrital hematite. No additional rock magnetic analyses were performed.

The majority of sites carried a reversed direction (following the polarity convention of Elston et al. 2022). Normal directions were found at the base of the formation at Poorman Creek, as well as in Flint Creek (lower) and Trail Gulch.  However, the rare "purple" beds exposed at Flint Creek (lower) and Trail Gulch were discarded due to concerns about rotation and/or post-depositional overprinting.  These remanences had relatively steep inclinations and appeared to be CW-rotated based on the paleomagnetic declinations.  No reversal test was reported from the remaining normal polarity site (see below). 

Sites pass a fold test with 100 $\pm$ 6% untilting as the optimum, using Watson and Enkin, 1993.

## Geochronology Constraints

Constenius et al. (accepted) did ID-TIMS U-Pb dating on baddeleyite from the lower Purcell Lava and report an age of 1438.9 ± 3.2 Ma.  Dating of instrusive sills using LA-ICP-MS corroborate magmatism at this age in the Belt Supergroup. The Purcell Lava intrudes at the contact of the Snowslip and the overlying Shepard Formation.

A minimum age constraint of 1454 ± 9 Ma from SHRIMP U-Pb zircon dating is given by the overlying metabentonite from the upper Helena Formation (Evans et al., 2000). A recent LA-ICP-MS U-Pb zircon date provides a quite similar age of 1450.4 ± 7.8 Ma (Constenius et al. accepted; this is a revision of a prior age published in Hirtz et al. 2024); these overlap substantially, and we use the Evans et al. (2000) date.  The Snowslip Formation directly overlies the Helena Formation in Glacier National Park, where this sample was collected.

Because of this tight age-bracketing of the Snowslip Formation, we choose to pick the midpoint between the two age constraints (1458 and 1435.7 Ma) to arrive at 1449 Ma.

## Get bedding from tilt correction to do fold test

```python
# Find sites that have both geographic and tilt-corrected coordinates

sites_geolist = []
for site in sites_geo["site"]:
    if sites_tilt["site"].str.contains(site, regex=False).any():
        sites_geolist.append(site)

decs_fold = []
incs_fold = []
dip_dirs = []
dips = []

for site in sites_geolist:
    dec_geo = float(sites_geo.loc[sites_geo["site"] == site, "dir_dec"].iloc[0])
    inc_geo = float(sites_geo.loc[sites_geo["site"] == site, "dir_inc"].iloc[0])
    dec_tilt = float(sites_tilt.loc[sites_tilt["site"] == site, "dir_dec"].iloc[0])
    inc_tilt = float(sites_tilt.loc[sites_tilt["site"] == site, "dir_inc"].iloc[0])

    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)

    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)

diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)
```

```python
ipmag.bootstrap_fold_test(diddd)
```

All sites measured by Elston et al. (2002) pass a fold test.  

We also performed a fold test on the subset included in the mean pole, reported in Elston et al. (2002) Table 3.

```python
sites_subset_geo = sites_geo[sites_geo['result_quality'] == 'g']

sites_subset_geolist = []

for site in sites_subset_geo['site']:
    if sites_tilt['site'].str.contains(site).any():
        sites_subset_geolist.append(site)

decs_fold = []
incs_fold = []        
dip_dirs = []
dips = []
for site in sites_subset_geolist:
    dec_geo = float(sites_geo[sites_geo['site'] == site]['dir_dec'])
    inc_geo = float(sites_geo[sites_geo['site'] == site]['dir_inc'])
    dec_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_dec'])
    inc_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_inc'])
    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)
    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)
    
diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)

ipmag.bootstrap_fold_test(diddd)    
```

Elston's subset also passes a fold test.
