# Western Channel diabase ca. 1590 Ma paleomagnetic pole

## Geologic context

The Western Channel diabase intrudes the folded Paleoproterozoic Rocks of the Cameron Bay and Echo Bay groups (Great Bear Magmatic Zone of Wopmay), and the generally flatlying Hornby Bay Group strata. The diabase intrudes as sheetlike bodies and dikes trending E and NNE. 

Irving et al. (1972) conducted paleomagnetic investigation of 26 sites with both drill (5-6 samples/site) and handsample of (3 samples/site). Hamilton and Buchan (2010) suggest that by combining the dyke and sheets sites adequately averages out the secular variation, yet they argue that there might only be two distinct VGPs from the dykes, Irving et al acknowledged this detail, by automatically increasing the reported pole A95. 
    
## Paleomagnetic and rock magnetic data

Magnetizations were made using a spinner magnetometer, rock magnetism includes MDF of ARM, IRM. Site 17 was the only one that was unstable. 

All samples were subjected to stepwise AF demagnetization; select samples underwent thermal demagnetization (baked contact samples and Port Radium diabase site 144). High coercivity components interpreted as primary displayed unblocking teperatures between 500-560°C consistent with magnetite as the primary carrier for the diabases.  No zjerderfeld diagrams were shown, or PCA mentioned, but the primary nature is supported by a positive baked-contact test, including fine grained sediments and porphyritic lavas, spanning 0.5 m to 60 m from the intrusions. The dataset is consistent with intrusions remagnetizatizing contact rocks <8 m from intrusions, while sites >1 km from contacts remained apparently unaffected by the intrusions.    Data was combined from 35 sites, resulting in a direction of 356°/-50° (a95=6), and pole of 9°N, 115°W (A95 = 6.0°, N = 35)

## Age constraints
-FA66-42 1590.2 ± 3.8 Ma (U–Pb ID-TIMS baddeleyite single grain weighted average Pb/Pb age; Hamilton and Buchan, 2010).    
-FA66-44 1591.9 ± 2.9 Ma (U–Pb ID-TIMS baddeleyite single grain weighted average Pb/Pb age; Hamilton and Buchan, 2010).    
-FA66045-6 1592.4 ± 2.5 Ma (U–Pb ID-TIMS baddeleyite fractions upper intercept age; Rogers et al., 2018)
