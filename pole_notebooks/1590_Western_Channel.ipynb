# Western Channel diabase ca. 1590 Ma paleomagnetic pole

## Geologic context

The Western Channel diabase intrudes the folded Paleoproterozoic Rocks of the Cameron Bay and Echo Bay groups (Great Bear Magmatic Zone of Wopmay), and the generally flatlying Hornby Bay Group strata. The diabase intrudes as sheetlike bodies and dikes trending E and NNE. 

Irving et al. (1972) conducted paleomagnetic investigation of 26 sites with both drill (5-6 samples/site) and handsample of (3 samples/site). Hamilton and Buchan (2010) suggest that by combining the dyke and sheets sites adequately averages out the secular variation, yet they argue that there might only be two distinct VGPs from the dykes, Irving et al acknowledged this detail, by automatically increasing the reported pole A95. 
    
## Paleomagnetic and rock magnetic data

Magnetizations were made using a spinner magnetometer, rock magnetism includes MDF of ARM, IRM. Site 17 was the only one that was unstable. 

All samples were subjected to stepwise AF demagnetization; select samples underwent thermal demagnetization (baked contact samples and Port Radium diabase site 144). High coercivity components interpreted as primary displayed unblocking teperatures between 500-560°C consistent with magnetite as the primary carrier for the diabases.  No zjerderfeld diagrams were shown, or PCA mentioned, but the primary nature is supported by a positive baked-contact test, including fine grained sediments and porphyritic lavas, spanning 0.5 m to 60 m from the intrusions. The dataset is consistent with intrusions remagnetizatizing contact rocks <8 m from intrusions, while sites >1 km from contacts remained apparently unaffected by the intrusions.    Data was combined from 35 sites, resulting in a direction of 356°/-50° (a95=6), and pole of 9°N, 115°W (A95 = 6.0°, N = 35)

## Age constraints
-FA66-42 1590.2 ± 3.8 Ma (U–Pb ID-TIMS baddeleyite single grain weighted average Pb/Pb age; Hamilton and Buchan, 2010).

## R2: Techniques and statistical analysis (Meert et al., 2020)

The R2 assessment follows the three sub-criteria of Meert et al. (2020) (`resources/Meert2020_R_criteria.md`): **(a)** stepwise demagnetization by at least two methods; **(b)** component analysis by PCA or great circles; and **(c)** adequate paleosecular-variation (PSV) sampling. Irving et al. (1972) isolated a stable "cleaned" remanence by stepwise alternating-field demagnetization (with thermal checks); (a) and (b) are reported below only as *advisory* indicators read from the MagIC `method_codes`. The substantive, reproducible part of R2 is sub-criterion (c), evaluated from the recreated site data with the Deenen et al. (2011) A95 envelope on the VGP distribution together with N ≥ 25 samples, 10 ≤ K ≤ 70, B ≥ 8 sites, and ≥ 3 samples per site.

```python
r2 = pt.assess_R2(sites_geo, pole_mean)
print()
pt.Deenen_test(pole_mean['n'], pole_mean['alpha95'])
pt.plot_Deenen_test(pole_mean)
plt.show()
```

### Supplementary: VGP distribution-shape diagnostics (Fisher Q-Q and SVEI)

Beyond the formal R2(c) thresholds above, these tests examine the *shape* of the VGP distribution as a further check on PSV sampling. The Fisher Q-Q test (`ipmag.fishqq`) compares the VGP distribution against a Fisher model in its longitude (uniform) and latitude (exponential) components. The SVEI test (`pmagpy.svei`) compares the elongation (E) and elongation azimuth (V2dec) of the directional distribution at the study locality against Monte Carlo realizations of a giant-Gaussian-process field model (TK03.GAD; Tauxe & Kent, 2004) at the implied paleolatitude. These are reported as diagnostics, not as a strict reliability cutoff.

```python
fishqq_result = pt.fishqq_vgps(sites_geo, unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(sites_geo, study_lon, study_lat, model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}\u00b0")
print(f"elongation E = {svei_result['E']:.2f} "
      f"(model 95% bounds {svei_result['Esim_min']:.2f}\u2013{svei_result['Esim_max']:.2f}; "
      f"{'consistent' if svei_result['E_result'] else 'inconsistent'})")
print(f"V2 declination = {svei_result['V2dec']:.1f}\u00b0 "
      f"(model 95% bounds {svei_result['V2sim_min']:.1f}\u2013{svei_result['V2sim_max']:.1f}; "
      f"{'consistent' if svei_result['V2_result'] else 'inconsistent'})")
```

The Deenen et al. (2011) A95 envelope is satisfied (A95 = 4.3° within [2.9°, 8.7°]), and N (130), K (32), and B (35) all pass; the Fisher Q-Q test finds the VGP distribution consistent with a Fisher distribution (the uniform-longitude Mu and exponential-latitude Me statistics both fall below their critical values). R2(c) nonetheless scores **0** because a few sites rest on only two samples (< 3 per site). The SVEI test, at the implied paleolatitude (~31°), returns an elongation magnitude (E ≈ 5.0) well above the TK03.GAD 95% range (≈1.1–3.3): the directional distribution is more elongated than secular variation alone would produce, most likely because the shallow-inclination Western Channel **dike** sites are pooled with the steeper diabase-**sheet** sites. The elongation azimuth (V2dec) is consistent with the model.
