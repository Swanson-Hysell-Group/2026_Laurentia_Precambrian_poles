# McNamara pole ~1390 Ma Elston et al. (2002) 

## Geologic context

The Mesoproterozoic Belt Supergroup, located in Montana and Idaho, USA and Alberta and British Columbia, Canada, spans an enormous extent covering > 200,000 km<sup>2</sup> (Lonn et al. 2020).  It is a predominantly sedimentary succession over 15 km thick in places, which has been subdivided into four groups. The McNamara Formation is in the Missoula Group, which is the uppermost group (Lonn et al. 2020).  The Belt Supergroup was an early place of paleomagnetic analyses in the 1950s and 1960s, but the majority of subsequent paleomagnetic analyses were the product of years of work by Don Elston of the United States Geologic Survey and colleagues, which were published in Elston et al. (2002).  While Elston et al. (2002) note that carbonates and green-grey-black siliclastics were analyzed, the "weak and incoherent magnetizations" led the study to focus on red beds spanning the entirety of the stratigraphic succession.

The McNamara Formation has a type section near Missoula, MT, where it is a red and green mudstone that is noted to be "chert-bearing", capped by a red quartz-rich sandstone unit with both flat-laminations and cross-bedding (Winston and Link 1993). Due to the microlaminae (mm-scale) and cm-scale laminae of siltstone and claystone, which contain abundant mudcracks, it is proposed to have been deposited in a playa mudflat environment that was covered by an alluvial environment. To the northwest, the McNamara fines and grades into green argillite, which is called the Libby Formation and -- depending on the study -- may include units of the Garnet Range Formation. The name McNamara is preferred by some across the entire basin within the United States, based on original publication age (Winston and Link 1993, Clapp and Deiss, 1931), while others use different names from east to west (Lonn et al. 2020). To the north in Canada, the McNamara Formation is correlated to the Roosville Formation (Lonn et al. 2020). 

## Paleomagnetic and rock magnetic data

According to Elston at el. (2002), "red beds are found near the base and top of the McNamara Formation with gray beds in between." No formal rock magnetic analyses were performed, but stepwise thermal demagnetizatin on select samples showed that a large component of the signal unblocked between 650 and 670/680 ºC, suggesting coarse-grained "specular" detrital hematite.   It is worth noting that the blanket demagnetization steps used to find paleomagnetic directions often did not reach these temperatures, with 5 sites at 590 ºC, two sites at 550 ºC, and two sites at 650 ºC.  It is noted that stepwise thermal demagnetization was performed on pilot samples in up to 8 steps.

Site mean directions are reported using Fisher statistics. Mean site directions in the McNamara Formation have smeared declinations but constant inclination, which was interpreted to indicate relative vertical-axis rotation of > 50º; however, since sites spanned a large area, rotations were assumed as random and a mean was taken of all the sites (with the exclusion of the reversed interval in Trout Creek due to "contamination"). The formation mean for 10 sites was published as a virtual geomagnetic pole (prior literature data from Collinson and Runcorn, 1960, was reported but not included in this mean).

The upper half of the McNamara is predominantly reverse polarity (as defined by Elston), contrasting with the normal polarity lower McNamara. Three sites showed mixed polarity in between these two polarity zones (West Twin Creek, Warm Spring Canyon, and the top of the Spotted Bear section, overlying 200 m of normal polarity). Reported site directions for Warm Springs Canyon and West Twin Creek were presumably flipped to the same hemisphere, such that the direction is reversed, while Spotted Bear is reported as fully normal (suggesting that the mixed polarity zone samples were not included). A further site (Trout Creek) is separated by polarity, with the reversed directions excluded. The initial paper suggested that normal and reversed directions were close to antiparallel, but no formal statistical analyses were reported. See below for formal statistical analyses.

The data did pass a fold test with 97 $\pm$ 7% untilting as the optimum using Watson and Enkin, 1993.

## Geochronology Constraints

Evans et al. (2000) reported a SHRIMP U–Pb zircon age of 1401 ± 6 Ma for a tuff in the very thin transition zone between the Bonner Quartzite and Libby Formation.  As noted above, this contact would correlate to the base of the McNamara Formation.  

Detrital zircon data from within the Roosville Formation finds maximum depositional ages of 
1375.4 ± 10.5 Ma (YC2σ[3+], MSWD = 4.20, n = 3) and 1386.3 ± 11.2 Ma (YC1σ[2+], MSWD = 1.32, n = 2) (Hirtz et al., 2024), suggesting the McNamara Formation may have had a slow sedimentation rate.

These sedimentary ages are consistent with an uppermost age constraint from a diabase in the Salmon River Arch of Idaho, which intrudes the Lemhi group (potentially the Yellowjacket Formation, although the stratigraphy has been substantially revised) and has an age of 1379 ± 1 Ma based on U-Pb TIMS dating of zircons (Doughty and Chamberlain 1996).  The Lemhi Group (and Yellowjacket Formation) has been correlated to units stratigraphically below the McNamara Formation within the Missoula Group, but has historically been used to mark the "end" of deposition within the Belt Basin.

We have chosen to focus on the igneous age constraints and split the difference between the two for a nominal age of 1490 Ma.

## Recalculate pole based on site-level VGPs

We have recalculated the VGP of Elston using individual site VGPs. We additionally have recalculated the pole for all sites and for including the interpreted "contaminated" site; we did not include the Collinson & Runcorn (1960) site.

```python
# Elston's subset
sites_subset = sites_tilt[sites_tilt['result_quality'] == 'g']
di_block_subset = ipmag.make_di_block(sites_subset['vgp_lon'].tolist(), sites_subset['vgp_lat'].tolist())
di_block_subset_flipped = pmag.flip(di_block_subset, combine=True)
decs_subset_flipped = [row[0] for row in di_block_subset_flipped]
incs_subset_flipped = [row[1] for row in di_block_subset_flipped]

pole_subset = ipmag.fisher_mean(dec=decs_subset_flipped, inc=incs_subset_flipped)

map_axis = ipmag.make_orthographic_map(200, 0, figsize=(5,5))

ipmag.plot_pole(map_axis, plon=pole_subset['dec']+180, 
                   plat=-pole_subset['inc'], 
                   A95=pole_subset['alpha95'], 
                   color = 'blue', marker='s', label='Elston+ sites')


# All including Collinson and Runcorn (1960)
di_block_all = ipmag.make_di_block(sites_tilt['vgp_lon'].tolist(), sites_tilt['vgp_lat'].tolist())
di_block_all_flipped = pmag.flip(di_block_all, combine=True)
decs_all_flipped = [row[0] for row in di_block_all_flipped]
incs_all_flipped = [row[1] for row in di_block_all_flipped]

pole_all = ipmag.fisher_mean(dec=decs_all_flipped, inc=incs_all_flipped)

ipmag.plot_pole(map_axis, plon=pole_all['dec']+180, 
                   plat=-pole_all['inc'], 
                   A95=pole_all['alpha95'], 
                   color = 'red', marker='s', label='All sites')

# All excluding Collinson and Runcorn (1960)
sites_subset2 = sites_tilt[(sites_tilt['result_quality'] == 'g') | (sites_tilt['site'] == 'Trout Creek R')]
di_block_subset2 = ipmag.make_di_block(sites_subset2['vgp_lon'].tolist(), sites_subset2['vgp_lat'].tolist())
di_block_subset2_flipped = pmag.flip(di_block_subset2, combine=True)
decs_subset2_flipped = [row[0] for row in di_block_subset2_flipped]
incs_subset2_flipped = [row[1] for row in di_block_subset2_flipped]

pole_subset2 = ipmag.fisher_mean(dec=decs_subset2_flipped, inc=incs_subset2_flipped)

ipmag.plot_pole(map_axis, plon=pole_subset2['dec']+180, 
                   plat=-pole_subset2['inc'], 
                   A95=pole_subset2['alpha95'], 
                   color = 'green', marker='s', label='Subset of sites')

plt.legend(loc='best')
```

Blue calculated pole is very close to the red one; green is an option that excludes Collinson and Runcorn (1960), but includes both data sets from Trout Creek as separate sites.  The last option includes the Trout Creek site twice and therefore weights that site proportionally more. Thus, we suggest simply using the subset chosen by Elston.

## Get bedding from tilt correction to do fold test

Note that data from Collinson and Runcorn (1960) is not included here since no geographic coordinates or bedding orientations are provided.

```python
# Find sites that have both geographic and tilt-corrected coordinates

sites_geolist = []
for site in sites_geo['site']:
    if sites_tilt['site'].str.contains(site).any():
        sites_geolist.append(site)

decs_fold = []
incs_fold = []        
dip_dirs = []
dips = []
for site in sites_geolist:
    dec_geo = float(sites_geo[sites_geo['site'] == site]['dir_dec'])
    inc_geo = float(sites_geo[sites_geo['site'] == site]['dir_inc'])
    dec_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_dec'])
    inc_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_inc'])
    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)
    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)
    
diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)
```

```python
ipmag.bootstrap_fold_test(diddd)
```

The data thus passes a fold-test.  

Bedding dips range from 9° to 45° and the samples are scattered across multiple separate fault panels/blocks.  Regional folding and faulting definitely occured in the Cretaceous, but it is unclear if earlier events occurred.  Many workers suggest synsedimentary normal faulting in the lower Belt Supergroup units. There is also the proposed Kootenay orogeny (1370–1300 Ma), as well as widespread metamorphism documented in the Belt Supergroup, with ages around 1100 Ma (Lonn et al. 2020).

```python
decs_trout = sites_tilt[sites_tilt['site'].str.contains('Trout')]['dir_dec'].tolist()
incs_trout = sites_tilt[sites_tilt['site'].str.contains('Trout')]['dir_inc'].tolist()
a95s_trout = sites_tilt[sites_tilt['site'].str.contains('Trout')]['dir_alpha95'].tolist()

plt.figure(figsize = (4,4))
ipmag.plot_net()

for i in range(len(incs_trout)):
    if incs_trout[i] > 0:
        ipmag.plot_di_mean(dec=decs_trout[i]-180, inc=-incs_trout[i], a95=a95s_trout[i])
    elif incs_trout[i] <= 0:
        ipmag.plot_di_mean(dec=decs_trout[i], inc=incs_trout[i], a95=a95s_trout[i])
```
