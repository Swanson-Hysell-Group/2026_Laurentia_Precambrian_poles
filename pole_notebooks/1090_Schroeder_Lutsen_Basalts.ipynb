# Schroeder-Lutsen basalts ca. 1090 Ma paleomagnetic pole

## Geologic context

The Schroeder-Lutsen basalts are a sequence of subophitic to ophitic olivine-tholeiite basalt flows that are the uppermost unit of the North Shore Volcanic Group exposed along the north shore of Lake Superior in Minnesota (Green et al., 2011). On the south end of their exposure, the basalts are mapped in slight angular unconformity with the upper southwest sequenc of the North Shore Volcanic Group, with the polymictic Little Marais conglomerate occurring above the unconformity in isolated regions (Miller and Chandler, 1997; Green et al., 2011). The simple structure of the upper southwest sequence of the North Shore Volcanic Group below this unconformity is strongly disturbed by the hypabyssal intrusions of the Beaver Bay Complex, whereas the Schroeder-Lutsen basalts remain undisturbed and were not penetrated by the intrusions. Dates from the Beaver Bay Complex therefore provide maximum age constraints on the basalts.

Fairchild et al. (2017) measured a stratigraphic section along the Two Island River near Schroeder, Minnesota, identifying 44 flows within 444 m of section and sampling 40 of them as paleomagnetic sites. The sampled flows range in observed thickness from ~1 to 27 m. Pipe vesicles are common at flow bases, flow interiors have ophitic to subophitic textures, and well-exposed flow tops display pahoehoe textures; vesiculated sections are amygdaloidal. Oxidation is apparent in red hematite staining of the rocks. Thin clastic dikes were observed within seven flows. Along the nearby Cross River, 3 km to the northeast, John Green observed a comparable 44 flows through ~460 m of section (Green, 1992).

## Paleomagnetic and rock magnetic data

Thermal demagnetization of the Schroeder-Lutsen basalt specimens reveals simple, single-component magnetizations isolated by principal-component analysis. Magnetization unblocks both through temperatures consistent with a (titano)magnetite-held remanence (up to ~580 °C) and through higher temperatures consistent with a hematite-held remanence (>580 °C). In many specimens significant remanence unblocks between 580 °C and 625 °C, suggesting a maghemite-held, chemically modified thermal remanence. Where a specimen carries remanence held by both mineralogies the directions are indistinguishable — the average difference between fits to the ~400–580 °C and 590–680 °C portions of the demagnetization spectra is only ~2.9° (Fairchild et al., 2017). Site means are westerly with moderate positive inclination and are of a single (normal, Portage Lake) polarity.

The pole combines the 40 new Two Island River flow sites of Fairchild et al. (2017) — magnetite (low-temperature) component, tilt-corrected — with 10 Schroeder-Lutsen basalt flow sites of Tauxe & Kodama (2009), for a Fisher mean of N = 50 site VGPs. The older single-step Books (1972) Schroeder data are treated separately and not included, following the published selection. The resulting pole at 27.1°N, 187.8°E (A95 = 3.0°, N = 50) is distinct from the pole of the underlying North Shore Volcanic Group.

A notable feature of the data is that the VGPs are irregularly clustered into two distinct populations — one more northerly and one more southerly — so that the overall VGP population is not consistent with a Fisherian distribution (Fisher Q-Q test; Fairchild et al., 2017). The two populations encompass the full range of rock magnetic behavior and recur at repeated stratigraphic levels, so the less common southerly population does not appear to reflect transient excursional behavior, and the consistent structural orientation of the flows through the section rules out faulting or differential deformation as the cause of the VGP offset. The clustering is most plausibly a record of secular-variation behavior of the late-stage rift geomagnetic field; the same irregular clustering is seen in the coeval Lake Shore Traps. Despite the non-Fisherian distribution, the A95 of the mean pole overlaps those of the Portage Lake Volcanics and the Lake Shore Traps, and its position is consistent with the Schroeder-Lutsen basalts being temporally between those two units.

## Age constraints

The Schroeder-Lutsen basalts are not directly dated. They postdate the 1091.61 ± 0.14 Ma (<sup>206</sup>Pb/<sup>238</sup>U, CA-ID-TIMS zircon) Silver Bay aplite dike of the Beaver Bay Complex, which Fairchild et al. (2017) collected just west of Palisade Head from the granophyre zone of one of the youngest Beaver Bay Complex intrusions and take as the best maximum age constraint for the basalts. They are likely older than the youngest dated Midcontinent Rift volcanism — the Michipicoten Island Formation (1083.52 ± 0.23 Ma). The maximum-age constraint of <1091.61 Ma is more consistent with assignment to the late stage of rift magmatism than with the earlier interpretation correlating the basalts with the base of the Portage Lake Volcanics, and indicates that olivine-tholeiitic basaltic volcanism continued as late as ca. 1091 Ma or younger. The pole is treated here as ca. 1090 Ma, bracketed between 1083.5 and 1091.6 Ma.

## R2: Techniques and statistical analysis (Meert et al., 2020)

R2(c), the reproducible PSV-adequacy sub-criterion, is scored from the recreated site data against the Deenen et al. (2011) A95 envelope; R2(a)/(b) are advisory indicators. Fairchild et al. (2017) used both AF and thermal demagnetization with principal-component analysis, so (a) and (b) are satisfied.

```python
r2 = pt.assess_R2(sites_tc, pole_mean)
print()
pt.Deenen_test(pole_mean['n'], pole_mean['alpha95'])
_ = pt.plot_Deenen_test(pole_mean)
```

### Supplementary: VGP distribution-shape diagnostics (Fisher Q-Q and SVEI)

These diagnostics characterize the shape of the VGP distribution. As reported by Fairchild et al. (2017), the Schroeder-Lutsen basalt VGPs are **non-Fisherian**: they form two irregular clusters (a more northerly and a more southerly population) rather than a single symmetric distribution. The Fisher Q-Q test is therefore expected to flag a non-uniform longitude / non-exponential colatitude. This is a real feature of the late-stage rift field (the same clustering appears in the Lake Shore Traps), not a sampling artifact — the populations span the full range of rock magnetic behavior and recur at repeated stratigraphic levels. The mean pole nonetheless provides a robust late-stage paleogeographic constraint, but the non-Fisherian scatter is reported here rather than smoothed over.

```python
fishqq_result = pt.fishqq_vgps(sites_tc, unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(sites_tc, study_lon, study_lat, model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}°")
print(f"elongation E = {svei_result['E']:.2f} "
      f"(model 95% bounds {svei_result['Esim_min']:.2f}–{svei_result['Esim_max']:.2f}; "
      f"{'consistent' if svei_result['E_result'] else 'inconsistent'})")
print(f"V2 declination = {svei_result['V2dec']:.1f}° "
      f"(model 95% bounds {svei_result['V2sim_min']:.1f}–{svei_result['V2sim_max']:.1f}; "
      f"{'consistent' if svei_result['V2_result'] else 'inconsistent'})")
```

## Nordic workshop ranking

Prior Nordic Workshop compilations assign the Schroeder-Lutsen basalts an 'A' ranking. The pole's precise maximum-age constraint, large number of flow sites (N = 50), tight A95 (3.0°), and clean rock-magnetic characterization make it a key late-stage Keweenawan Track tie point despite an R-score of 5/7 (the failing criteria are R4, no field test, and R6, single polarity). Recreating the pole from the site level following the selection of Fairchild et al. (2017) (40 Two Island River flows + 10 Tauxe & Kodama flows; N = 50) reproduces their published pole (27.1°N, 187.8°E) and supports retaining 'A'. The non-Fisherian, two-cluster VGP distribution is reported as a real feature of the late-stage rift field rather than treated as a defect of the pole.
