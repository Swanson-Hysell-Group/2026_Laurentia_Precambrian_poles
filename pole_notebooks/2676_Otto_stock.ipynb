## Geologic Context

- Terrane: Laurentia-Superior(East).
- Compiled unit/pole label: Otto Stock Dykes and Aureole.

## Paleomagnetic And Rock Magnetic Data

- B = 18 sites, N = 84 samples; mean direction DEC/INC = 330.0°/71.0°; K = 139.0.
- Component note: N+R Component.
- Field/statistical tests noted in compilation: C*+,R-.
- Rock-magnetic interpretation: expand with mineral carriers/coercivity/unblocking evidence from source studies.

## Age Constraints

- Current age assignment: nominal age: 2676 Ma, 2671 Ma, 2681 Ma; reference method: 2680±1 for stock, but lamprophyre dykes are younger (ca.2670-2675), J.Ayers, pers.comm. 2008
- **Proposed bracket: nominal age: 2680 Ma, lomagage 2679.0 Ma, himagage 2681.0 Ma.**
- Compilation method note: The lamprophyre age has never been published but their emplacement was interpreted to be late stage intrusions associated with the stock. Let's put the current age assignment to be the age of the Otto Stock itself given that it is the best published age constraint on the magmatism.
