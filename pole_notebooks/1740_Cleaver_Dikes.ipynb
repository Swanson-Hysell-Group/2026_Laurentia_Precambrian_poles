# Cleaver Dykes ca. 1740 Ma paleomagnetic pole

## Geologic context

The Cleaver Dykes are a swarm of subvertical, post-orogenic diabase dykes in the
Great Bear magmatic zone of the Hottah terrane (Wopmay Orogen). They trend east–west to
east-southeast–west-northwest, are typically 5–20 m wide with sharp contacts, and
are sub-ophitic to ophitic with euhedral labradorite phenocrysts (Hildebrand,
1984). They are the oldest post-orogenic intrusions of the region, cutting the
gently tilted and folded but weakly metamorphosed ca. 1875 Ma Labine Group (U-Pb 1876.4 ± 2.4 and 1872.6 ± 0.9 Ma; Ootes et al., 2015)
and the plutons of the Great Bear arc. Younger intrusions in turn cut the Cleaver
Dykes: the ca. 1590 Ma Western Channel Diabase and the ca. 780 Ma Hottah
(Gunbarrel) Gabbro sheets (779.6 ± 1.4 Ma; Harlan et al., 2003). Dykes were sampled around Conjuror Bay in the
southeast corner of Great Bear Lake over an area of about 15 by 25 km, with one
additional dyke (the Echo Bay re-sampling) 50 km to the north.

## Paleomagnetic and rock magnetic data

Oriented cores (usually six per site) were collected from 32 dykes. Fifteen dykes
carried magnetizations scattered about the present field or interpreted to have been overprinted by
lightning-induced isothermal remanence and gave no useful information; seventeen
dykes carried stable magnetizations. Specimens were demagnetized by both
alternating-field and thermal methods, and characteristic directions were isolated
by principal-component line fitting. Thermal demagnetization is commonly
square-shouldered with unblocking temperatures predominantly between 500 and
600 °C, indicating a remanence carried by magnetite. The Cleaver
magnetization is directed to the southeast with intermediate downward inclination
and is of a single ("normal") polarity, with a mean direction of D = 136.5°,
I = 57.4° (k = 64, α95 = 4.5°, N = 17 dykes).

Bringing the 17 dyke site VGPs to a common polarity and computing the Fisher mean
gives a pole at 20.4°N, 274.9°E (A95 = 7.0°, K = 27.1, N = 17). This calculated poles position is similar to, but slightly different than, the
published Irving et al. (2004) Cleaver Dykes pole (19.4°N, 276.7°E, A95 = 6.1°)

A primary,
emplacement-age origin for the Cleaver magnetization is supported by the fine-grained
magnetite carrier, a positive baked-contact test where the dykes bake the adjacent
Labine Group, and a positive inverse baked-contact test in which the younger Western
Channel Diabase and Hottah Gabbro that cut the dykes carry different directions.

## Age constraints

- 1740 +5/−4 Ma (U-Pb baddeleyite; Irving et al., 2004). A modified York regression
  of four baddeleyite fractions from a thick Cleaver dyke at Conjuror Bay (sample
  SSG-00-51) gives an upper-intercept age interpreted as the crystallization age of
  the dyke and, by their consistent field characteristics and magnetization
  directions, of the Cleaver swarm as a whole. The nominal age adopted here is
  1740 Ma (1736–1745 Ma).

## Comparison with the published pole

The pole recreated above from the 17 site VGPs (20.4°N, 274.9°E, A95 = 7.0°) lies ~1.9° from the pole published by Irving et al. (2004) (19.4°N, 276.7°E, A95 = 6.1°), well within their mutual confidence limits. Both are means of the same 17 dykes — the published α95 = 4.5° (k = 64, N = 17) is an N = 17 unit-mean statistic; a mean of all 213 specimens at that k would give α95 ≈ 1.2°, not 4.5°. The small offset has two sources, and rounding of the published table is not one of them (a Monte Carlo test holds k at 50.4 ± 0.05 under ±0.05° perturbation of the tabulated directions):

1. **How the pole is calculated (~0.9°).** Irving et al. converted a single *mean direction* to a pole, reporting an oval confidence limit (dp/dm = 6.1°). The recreation here instead takes the Fisher mean of the 17 individual site *VGPs*, which yields a circular A95 = 7.0°. Recomputing the recreation Irving's way (mean direction → pole) moves it to 275.6°E/19.8°N, closing most of this component.

2. **The input dyke-mean directions (~1.1°, the larger part).** Irving et al.'s grand mean (136.5°/57.4°, k = 64) is computed from each dyke's full set of specimen line fits (their Table 2, "the mean using all specimens"), whereas the recreation can only average the site-mean directions as tabulated in their Table 1 (137.8°/58.4°, k = 50.4). The tabulated site means differ by a few degrees from the published dyke means at the more scattered dykes (e.g. sites 6, 27, 37, which sit 15–21° from the mean), which both shifts the mean ~1° and raises the published k to 64. The precise reason the Table 1 directions differ from the directions entering their grand mean is not resolvable from the published tables alone.

Because the published pole depends on specimen-level line fits that do not accompany the contribution, it cannot be reproduced from the published site table alone. The values exported to the Nordic summary are therefore the **recalculated** pole and direction, which are exactly reproducible from the archived MagIC site data; the published pole is retained here for comparison.

```python
# recreated pole (this study, from the 17 site VGPs) vs. the published pole of
# Irving et al. (2004) (GPMDB 9139)
published_plat, published_plon, published_A95 = 19.4, 276.7, 6.1
recreated_plat, recreated_plon = pole_mean['inc'], pole_mean['dec']
separation = float(pmag.angle([recreated_plon, recreated_plat],
                              [published_plon, published_plat])[0])
print(f"recreated pole: {recreated_plat:.1f} N, {recreated_plon:.1f} E, "
      f"A95 {pole_mean['alpha95']:.1f}  (N={pole_mean['n']})")
print(f"published pole: {published_plat:.1f} N, {published_plon:.1f} E, "
      f"A95 {published_A95:.1f}  (Irving et al., 2004)")
print(f"angular separation: {separation:.1f} deg (within the mutual A95)")
```

## Field tests for a primary magnetization (R4)

Irving et al. (2004) used baked-contact relationships to show that the Cleaver
magnetization was acquired during initial cooling. The clearest example is at dyke
34: baked andesite of the Labine Group at the contact carries the same direction as
the dyke, whereas andesite 50 m away retains a distinct, pre-Cleaver direction (a
positive baked-contact test). A comparable test is recorded at dyke 22 against
Labine Group tuff. The test is reinforced in the inverse sense by the younger
intrusions that cut the dykes — the ca. 1590 Ma Western Channel Diabase and the
ca. 780 Ma Hottah (Gunbarrel) Gabbro (779.6 ± 1.4 Ma; Harlan et al., 2003) — whose directions differ from the Cleaver
magnetization, showing that the dykes were not overprinted by these later events.
The stereonet below shows the dyke-34 baked-contact test and the Hottah Gabbro
direction relative to the Cleaver Dykes mean.

```python
# load all sites (including the result_quality 'b' control rocks) for the field-test plot
sites_all, _ = pt.load_magic_sites('../data/1740_Cleaver_Dikes/sites.txt', drop_bad=False)

def site_dir(site):
    row = sites_all[sites_all['site'].astype(str) == site].iloc[0]
    return float(row['dir_dec']), float(row['dir_inc'])

d34 = site_dir('34')   # Cleaver dyke 34
s35 = site_dir('35')   # baked andesite at contact of dyke 34
s36 = site_dir('36')   # andesite 50 m from dyke 34 (unbaked)
g45 = site_dir('45')   # ca. 780 Ma Hottah Gabbro (younger intrusion)

ipmag.plot_net()
ipmag.plot_di(dir_mean['dec'], dir_mean['inc'], color='red', marker='*',
              markersize=120, label='Cleaver Dykes mean')
ipmag.plot_di(d34[0], d34[1], color='firebrick', marker='s', label='dyke 34')
ipmag.plot_di(s35[0], s35[1], color='darkorange', marker='o',
              label='baked andesite at contact (site 35)')
ipmag.plot_di(s36[0], s36[1], color='royalblue', marker='^',
              label='andesite 50 m away (site 36)')
ipmag.plot_di(g45[0], g45[1], color='purple', marker='D',
              label='Hottah Gabbro ca. 780 Ma (site 45)')
ipmag.plt.legend(loc='center left', bbox_to_anchor=(1.05, 0.5), fontsize=9)
ipmag.plt.title('Baked-contact test: dyke 34 and the Hottah Gabbro')
ipmag.plt.show()
```

## R2: Techniques and statistical analysis (Meert et al., 2020)

The R2 assessment follows the three sub-criteria of Meert et al. (2020), as set out
in `resources/Meert2020_R_criteria.md`: (a) stepwise demagnetization by at least two
methods (AF and thermal); (b) component analysis by PCA (best-fit lines) or great
circles; and (c) adequate paleosecular-variation (PSV) sampling. Sub-criteria (a) and
(b) are reported only as advisory indicators read from the MagIC `method_codes`; both
alternating-field and thermal demagnetization were applied and characteristic
directions were isolated by principal-component line fitting (Irving et al., 2004),
so both are met. The substantive, reproducible part of R2 is sub-criterion (c),
evaluated from the recreated site data with the Deenen et al. (2011) A95 envelope on
the VGP distribution together with N ≥ 25 samples, 10 ≤ K ≤ 70, B ≥ 8 sites, and
≥ 3 samples per site.

```python
r2 = pt.assess_R2(sites_pole, pole_mean)
print()
pt.Deenen_test(pole_mean['n'], pole_mean['alpha95'])
pt.plot_Deenen_test(pole_mean)
plt.show()
```

### Supplementary: VGP distribution-shape diagnostics (Fisher Q-Q and SVEI)

Beyond the formal R2(c) thresholds above, these tests examine the shape of the VGP
distribution as a further check on PSV sampling. The Fisher Q-Q test (`ipmag.fishqq`)
compares the VGP distribution against a Fisher model in its longitude (uniform) and
latitude (exponential) components. The SVEI test (`pmagpy.svei`) compares the
elongation (E) and elongation azimuth (V2dec) of the directional distribution at the
study locality against Monte Carlo realizations of a giant-Gaussian-process field
model (TK03.GAD; Tauxe & Kent, 2004) at the implied paleolatitude. These are reported
as diagnostics, not as a strict reliability cutoff.

```python
fishqq_result = pt.fishqq_vgps(sites_pole, unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(sites_pole, study_lon, study_lat, model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}°")
print(f"elongation E = {svei_result['E']:.2f} "
      f"(model 95% bounds {svei_result['Esim_min']:.2f}–{svei_result['Esim_max']:.2f}; "
      f"{'consistent' if svei_result['E_result'] else 'inconsistent'})")
print(f"V2 declination = {svei_result['V2dec']:.1f}° "
      f"(model 95% bounds {svei_result['V2sim_min']:.1f}–{svei_result['V2sim_max']:.1f}; "
      f"{'consistent' if svei_result['V2_result'] else 'inconsistent'})")
```

At N = 17 dykes the data sample paleosecular variation adequately by the formal
R2(c) thresholds (Deenen et al., 2011, envelope satisfied; K within 10–70). The
SVEI test finds the elongation and its azimuth consistent with the TK03.GAD field
model at the implied paleolatitude (~39°), though the modest sample size makes the
model bounds wide. These shape diagnostics are reported as supplements to R2(c)
rather than as pass/fail gates.
