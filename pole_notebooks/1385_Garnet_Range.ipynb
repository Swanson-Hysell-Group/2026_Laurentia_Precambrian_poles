# Garnet Range-Libby-Pilcher pole ~1390 Ma Elston et al. (2002)

## Geologic context

The Mesoproterozoic Belt Supergroup, located in Montana and Idaho, USA and Alberta and British Columbia, Canada, spans an enormous extent covering > 200,000 km2 (Lonn et al. 2020). It is a predominantly sedimentary succession over 15 km thick in places, which has been subdivided into four groups. The Garnet Range, Libby, and Pilcher Formations are at the top of the Missoula Group, which is the uppermost group (Lonn et al. 2020). The Belt Supergroup was an early place of paleomagnetic analyses in the 1950s and 1960s, but the majority of subsequent paleomagnetic analyses were the product of years of work by Don Elston of the United States Geologic Survey and colleagues, which were published in Elston et al. (2002). While Elston et al. (2002) note that carbonates and green-grey-black siliclastics were analyzed, the "weak and incoherent magnetizations" led the study to focus on red beds spanning the entirety of the stratigraphic succession.

The Garnet Range Formation, in its type area near Missoula, MT, sharply overlies the McNamara and green-grey micaceous quartz-rich sandstone and mudstone (Winston and Link 1993).  It marks a clear transition in the underlyin Belt-Purcell Supergroup units, with an abundance of fresh detrital minerals (feldspar and unweathered magnetite in addition to the mica) and lenses of sand with large-scale hummocky cross-stratification, interpreted as expansion of the Belt Basin (Winston and Link 1993).   The Garnet Range is included in the Libby Formation in western MT and Idaho and the Rooseville Formation in Canada (Lonn et al. 2020).  The Garnet Range is overlain in the Missoula Area by the Pilcher Formation, a purple and white cross-bedded sand that has a limited outcrop area (Winston and Link, 1993). It has been suggested to be the final filling of the Belt Basin by tidal/fluvial deposits.  

## Paleomagnetic and rock magnetic data

Elston et al. (2002) reports 5 sites--one from the Pilcher Formation, three from the Garnet range, and one from the Libby Formation (presumably the upper portion).  Elston et al. (2002) emphasize that these units were grouped to have enough localities.

The text notes that in most samples a >50%  magnetization is lost above 590 ºC suggesting hematite as an important remanence carrier.  Zijdervelds of two samples show a large drop in magnetization above 650 ºC highlighting the importance of coarse detrital hematite in these samples.  No additional rock magnetism was performed on these samples; microscopy of the unweathered detrital magnetite noted in the Garnet Range has not been published by prior authors.

The USGS subjected pilot samples to stepwise demagnetization in 8 steps, but then demagnetized the remaining samples in blanket demagnetization steps--all sites were treated this way, using steps of 550 ºC (1 Garnet Range site), 590 ºC (Pilcher and Libby sites), and 630 ºC (2 Garnet Range sites).  

Of the five sites, one was excluded form the formation mean direciton and pole--the Garnet Range Formation at Flint Creek--due to interpreted counterclockwise vertical-axis rotation based on paleomagnetic directions.  Elston et al. (2002) note caution with using this pole as the other four sites are small in number and near major structures that could also cause vertical-axis rotation.

All of the sites share the same polarity (reversed as per the nomenclature of this study.)

## Geochronology Constraints

Evans et al. (2000) reported a SHRIMP U–Pb zircon age of 1401 ± 6 Ma for a tuff in the very thin transition zone between the Bonner Quartzite and Libby Formation. 

Detrital zircon U-Pb dating from the Garnet Range and Pilcher Formations using LA-ICP-MS found some older MDAs, but also provided two constraints of 1384.0 ± 23.5 Ma and 1390.4 ± 24.9 Ma (Parker and Hendrix 2022).

These sedimentary ages are consistent with an uppermost age constraint from a diabase in the Salmon River Arch of Idaho, which intrudes the Lemhi group (potentially the Yellowjacket Formation, although the stratigraphy has been substantially revised) and has an age of 1379 ± 1 Ma based on U-Pb TIMS dating of zircons (Doughty and Chamberlain 1996). The Lemhi Group (and Yellowjacket Formation) has been correlated to units stratigraphically below the McNamara Formation within the Missoula Group, but has historically been used to mark the "end" of deposition within the Belt Basin.

Stratigraphic relationships show this unit is younger than the McNamara Formation (which forms the lower part of the Libby Formation)--we have thus chosen a nominal age of 1385 Ma.

## Recalculate pole

```python
# Elston's subset
sites_subset = sites_tilt[sites_tilt['result_quality'] == 'g']
di_block_subset = ipmag.make_di_block(sites_subset['vgp_lon'].tolist(), sites_subset['vgp_lat'].tolist())
di_block_subset_flipped = pmag.flip(di_block_subset, combine=True)
decs_subset_flipped = [row[0] for row in di_block_subset_flipped]
incs_subset_flipped = [row[1] for row in di_block_subset_flipped]

pole_subset = ipmag.fisher_mean(dec=decs_subset_flipped, inc=incs_subset_flipped)

map_axis = ipmag.make_orthographic_map(200, 0, figsize=(5,5))

ipmag.plot_pole(map_axis, plon=pole_subset['dec']+180, 
                   plat=-pole_subset['inc'], 
                   A95=pole_subset['alpha95'], 
                   color = 'blue', marker='s', label='Subset of sites')


# All
di_block_all = ipmag.make_di_block(sites_tilt['vgp_lon'].tolist(), sites_tilt['vgp_lat'].tolist())
di_block_all_flipped = pmag.flip(di_block_all, combine=True)
decs_all_flipped = [row[0] for row in di_block_all_flipped]
incs_all_flipped = [row[1] for row in di_block_all_flipped]

pole_all = ipmag.fisher_mean(dec=decs_all_flipped, inc=incs_all_flipped)

ipmag.plot_pole(map_axis, plon=pole_all['dec']+180, 
                   plat=-pole_all['inc'], 
                   A95=pole_all['alpha95'], 
                   color = 'red', marker='s', label='All sites')

plt.legend(loc='best')
```

Blue calculated pole is very close to the red one and visually the excluded site does look rotation. We suggest using the subset chosen by Elston et al. (2002).

# Get bedding from tilt correction to do fold test

```python
# Find sites that have both geographic and tilt-corrected coordinates

sites_geolist = []
for site in sites_geo['site']:
    if sites_tilt['site'].str.contains(site).any():
        sites_geolist.append(site)

decs_fold = []
incs_fold = []        
dip_dirs = []
dips = []
for site in sites_geolist:
    dec_geo = float(sites_geo[sites_geo['site'] == site]['dir_dec'])
    inc_geo = float(sites_geo[sites_geo['site'] == site]['dir_inc'])
    dec_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_dec'])
    inc_tilt = float(sites_tilt[sites_tilt['site'] == site]['dir_inc'])
    dip_dir, dip = pmag.get_tilt(dec_geo, inc_geo, dec_tilt, inc_tilt)
    decs_fold.append(dec_geo)
    incs_fold.append(inc_geo)
    dip_dirs.append(dip_dir)
    dips.append(dip)
    
diddd = ipmag.make_diddd_array(decs_fold, incs_fold, dip_dirs, dips)
    
```

```python
ipmag.bootstrap_fold_test(diddd)
```

The data thus passes a fold-test.  

Bedding dips range from 5.5° to 40° and the samples are scattered across separate fault panels/blocks.  Regional folding and faulting definitely occured in the Cretaceous, but it is unclear if earlier events occurred.  Many workers suggest synsedimentary normal faulting in the lower Belt Supergroup units. There is also the proposed Kootenay orogeny (1370–1300 Ma), as well as widespread metamorphism documented in the Belt Supergroup, with ages around 1100 Ma (Lonn et al. 2020).
