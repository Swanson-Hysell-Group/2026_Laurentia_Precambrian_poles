# Mamainse Point Volcanics ca. 1109–1094 Ma paleomagnetic poles

## Geologic context

The Mamainse Point Volcanics crop out along the eastern shore of Lake Superior in
Ontario, near where the covered southeast arm of the Midcontinent Rift meets the
lake. The succession of picritic to basaltic-andesite lava flows is interpreted
as the most continuous and best-exposed record of Keweenawan rift volcanism and
geomagnetic reversals, spanning much of the duration of rift magmatism
(Swanson-Hysell et al., 2014a). Unlike other extrusive successions of the rift, it
preserves three geomagnetic reversals; the polarity zones are geochemically
distinct and there is no geological evidence for fault repetition, supporting the
interpretation that the reversals are real rather than structural duplications.

In ascending stratigraphic order the succession records the Alona Bay
reversed-polarity zone, the Flour Bay normal-polarity zone, the Flour Bay
reversed-polarity zone, and — above the ~300 m Great Conglomerate — the Portage
Lake normal-polarity zone. The Great Conglomerate is interpreted to correspond to
the "latent stage" of reduced magmatic activity (ca. 1106–1100 Ma) that is
recorded as missing section in the western Lake Superior basin. A U-Pb date of
1100.36 ± 0.25 Ma on a tuff within the Flour Bay reversed zone, below the Great
Conglomerate, shows that eruptive activity continued at Mamainse Point during that
latent stage (Swanson-Hysell et al., 2014a).

The flows were sampled in stratigraphic context (Swanson-Hysell et al., 2009,
2014a); interflow paleohorizontal indicators allow structural (tilt) corrections
of the site mean directions. The progressive decrease in paleomagnetic
inclination up through the succession is interpreted to record rapid equatorward
plate motion during the late Mesoproterozoic assembly of Rodinia.

## Pole context

Following Swanson-Hysell et al. (2009, 2014a), poles are calculated from
stratigraphic subsets of the site VGPs, giving four poles presented from oldest to
youngest:

1. Mamainse lower R1 — ca. 1109 Ma; the lowermost ~600 m, the older part of the
   Alona Bay reversed-polarity zone. GPMDB 9510.
2. Mamainse lower R2 — ca. 1105 Ma; the younger part of the Alona Bay
   reversed-polarity zone. GPMDB 9511.
3. Mamainse Flour Bay — ca. 1100.36 Ma; the Flour Bay normal- and
   reversed-polarity zones combined (latent stage). GPMDB 9512.
4. Mamainse upper N — ca. 1094 Ma; the Portage Lake normal-polarity zone above the
   Great Conglomerate. GPMDB 9513.

All four are existing Nordic-compilation poles (GPMDB 9510–9513) and are exported
to the Nordic summaries. The Flour Bay pole combines a normal and a reversed zone,
whose VGPs are antipodal, so polarities are unified before averaging. Three
transitional flows (~1810–1860 m) recording the Flour Bay normal-to-reversed
reversal are excluded from the stable-polarity poles.

## Paleomagnetic and rock magnetic data

The characteristic remanence is carried by magnetite and isolated by stepwise
alternating-field and thermal demagnetization with principal-component analysis;
site means are tilt-corrected. A positive intraformational conglomerate test
(Swanson-Hysell et al., 2009) and positive reversal tests across the multiple
geomagnetic reversals (Swanson-Hysell et al., 2009, who found no asymmetry across
the reversals) support a primary origin for the magnetization.

## Age constraints

- A U-Pb (<sup>206</sup>Pb/<sup>238</sup>U, CA-ID-TIMS) date of 1100.36 ± 0.25 Ma
  on zircon from a tuff within the Flour Bay reversed zone (below the Great
  Conglomerate) directly dates the Flour Bay pole (Swanson-Hysell et al., 2014a).
- The Alona Bay reversed poles are older: ca. 1109 Ma (R1; 1106–1112 Ma) and ca.
  1105 Ma (R2; 1100.4–1109 Ma). The Portage Lake normal pole is ca. 1094 Ma
  (1090–1100 Ma). These ages follow the Midcontinent Rift age model of
  Swanson-Hysell et al. (2019).

## Site-level data and the four pole groups

Site means are loaded from `../data/1109_Mamainse_Point_Volcanics/sites.txt`,
assembled in `contribution_build/` from the Swanson-Hysell et al. (2014a) Mamainse
Point site table (MagIC contribution 16333), with the four poles written to
`locations.txt`. The `location` field tags each site with the pole it contributes
to; all site means are tilt-corrected.

```python
sites_geo, sites_tc = pt.load_magic_sites('../data/1109_Mamainse_Point_Volcanics/sites.txt')
print(f'{len(sites_tc)} tilt-corrected site means')
print(sites_tc['location'].value_counts())

LOCALITY = {'Mamainse lower R1':  (47.10, 275.29),
            'Mamainse lower R2':  (47.09, 275.27),
            'Mamainse Flour Bay': (47.07, 275.26),
            'Mamainse upper N':   (47.06, 275.25)}
results = {}
```

## Mamainse lower R1 — older Alona Bay reversed-polarity zone (ca. 1109 Ma)

The lowermost ~600 m of the succession; the oldest and most southerly Mamainse pole (here in the reversed convention, plotted at high north latitude after polarity unification), N = 24.

```python
sub = sites_tc[sites_tc['location'] == 'Mamainse lower R1'].reset_index(drop=True)
study_lat, study_lon = LOCALITY['Mamainse lower R1']
vgp_block, pole_mean = pt.compute_mean_pole(sub, unify_polarity=True)
ipmag.print_pole_mean(pole_mean)
_ = pt.plot_vgps_and_pole(vgp_block, pole_mean, central_longitude=pole_mean['dec'],
                          central_latitude=pole_mean['inc'], figsize=(5, 5))
```

```python
dir_block, dir_mean = pt.compute_mean_direction(sub, unify_polarity=True)
ipmag.print_direction_mean(dir_mean)
ipmag.plot_net()
ipmag.plot_di(di_block=dir_block, color='blue', marker='o')
ipmag.plot_di_mean(dir_mean['dec'], dir_mean['inc'], dir_mean['alpha95'], color='red', marker='s')
plt.title('Mamainse lower R1 — site mean directions (tilt-corrected)')
plt.show()
results['Mamainse lower R1'] = dict(sub=sub, pole=pole_mean, dir=dir_mean,
                        study_lat=study_lat, study_lon=study_lon)
```

```python
mp_r1 = results['Mamainse lower R1']
pt.Deenen_test(mp_r1['pole']['n'], mp_r1['pole']['alpha95'])
pt.plot_Deenen_test(mp_r1['pole'])
```

```python
# VGP distribution-shape diagnostics on the lowermost reversed pole
fishqq_result = pt.fishqq_vgps(mp_r1['sub'], unify_polarity=True)
fishqq_result
```

```python
svei_result = pt.svei_test_vgps(mp_r1['sub'], mp_r1['study_lon'], mp_r1['study_lat'],
                                model='TK03_GAD', plot=True)
print(f"paleolatitude = {svei_result['lat']:.1f}\u00b0; elongation E = {svei_result['E']:.2f} "
      f"({'consistent' if svei_result['E_result'] else 'inconsistent'} with TK03.GAD)")
```

## The four Mamainse poles in the context of the Laurentia APWP

```python
Laurentia_poles = pt.get_Laurentia_poles()
ax = pt.plot_apwp_context(Laurentia_poles, results['Mamainse lower R1']['pole']['inc'], results['Mamainse lower R1']['pole']['dec'], results['Mamainse lower R1']['pole']['alpha95'], age_min=635, age_max=1800, projection='orthographic', central_longitude=results['Mamainse lower R1']['pole']['dec'], central_latitude=results['Mamainse lower R1']['pole']['inc'])
for loc, color in [('Mamainse lower R2', 'firebrick'),
                   ('Mamainse Flour Bay', 'goldenrod'),
                   ('Mamainse upper N', 'darkorange')]:
    p = results[loc]['pole']
    ipmag.plot_pole(ax, p['dec'], p['inc'], p['alpha95'], color=color, markersize=40, label=loc)
ax.legend(loc='lower left')
plt.show()
```

The four poles trace a progression along the Keweenawan Track from high
(49.5°N, R1) to low (31.2°N, upper N) pole latitude — the progressive decrease in
paleomagnetic inclination through the succession that Swanson-Hysell et al. (2014a)
interpret as rapid equatorward plate motion of Laurentia.
